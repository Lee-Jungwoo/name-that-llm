## 1. 풀려는 문제

짧은 영어 지문 하나를 보고 **18개 LLM 중 어느 모델이 썼는지** 맞히는 18-클래스 텍스트 분류 문제입니다.

- **입력**: 지문 1개 (단어 36~274개, 중앙값 약 100개)
- **출력**: 모델 이름 1개 (GPT 5개, Gemini 4개, Claude 5개, Qwen·GLM·Kimi 4개)
- **지표**: balanced accuracy (클래스별 recall의 평균). 무작위로 찍으면 1/18 ≈ 0.056입니다.
- **어려운 점**: 테스트 지문의 프롬프트(주제)는 train에 한 번도 나오지 않습니다. 그래서 *무엇을* 썼는지가 아니라 **어떻게** 썼는지(문체)를 배워야 합니다. 같은 연구소 모델끼리는 문체가 비슷해서 서로 많이 헷갈립니다.

## 3. Model Architecture

Models with very different inductive biases are trained separately, and their calibrated log-probabilities are blended. The diagram below shows branches A and B; **v7 adds branch C (DeBERTa-v3-large), described in 3.5**, and blends all three with `LP = w_B * LP_B + w_C * LP_C + w_A * LP_A`.

```
                               text (one passage)
                                       |
              +------------------------+-------------------------+
              |                                                  |
              v                                                  v
 [A] n-gram branch (CPU)                          [B] LoRA branch (GPU)
 4 views of the same text:                        Qwen3-4B-Base tokenizer
   char  : raw text, char 3-5-grams                 ids [B,T], T <= 384
   word  : raw text, word 1-2-grams                        |
   shape : Hello -> X, the -> x                   token embedding   [B,T,2560]
   func  : keep function words only                        |
           (the _ of _)                           18 x Qwen3 decoder block (frozen)
              |                                     + LoRA (r=16) on q/k/v/o
   TF-IDF per view, weighted, hstack                       |
   + 27 stylometric stats (v5/v8, norm ~0.2)               |
   X [N, <=510k] (sparse)                         tap hidden states at depths
              |                                     2, 4, 6, 9, 12, 15, 18
   LinearSVC, one-vs-rest                         masked mean-pool + LayerNorm each
   S [N,18]                                       concat -> [B,17920]
              |                                            |
              |                                   linear head -> logits [B,18]
              |                                            |
   LP_A = log_softmax(S / T_A)                    LP_B = log_softmax(logits / T_B)
              |                                            |
              +-------------------+------------------------+
                                  |
                     LP = w * LP_B + (1 - w) * LP_A             [N,18]
                                  |
                prompt centering inside topic clusters           [N,18]
                                  |
                damped Sinkhorn: per-class log bias              [N,18]
                                  |
                          argmax over k -> predicted model
```

- **A. n-gram branch** captures surface habits: punctuation, spelling, sentence rhythm. The `shape` and `func` views delete topic words, so they are less tempted to memorize the prompt. Fast, deterministic baseline.
- **B. LoRA branch** reuses the stylistic representations a pretrained LM already has. Style information survives better in early/middle layers than in the last one, so only the first 18 of 36 layers are kept and several depths are pooled together. LoRA is implemented by hand (no `peft`) so the notebook runs offline. Trained on the Open Pool first (Stage 1), then fine-tuned on `train` (Stage 2).
- **Blending**: the two temperatures are chosen on out-of-fold predictions (2,699 passages; each LoRA fold model never saw its own validation prompts). `w` is fixed at 0.6. Without a GPU or Qwen3 weights, branch B is skipped (`w = 0`).
- **Prompt centering**: passages that answer the same prompt share a topic. Subtracting the cluster mean removes part of the topic signal and leaves the per-model style difference.
- **Class-ratio correction (Sinkhorn)**: a per-class bias is added so that the predicted class shares move toward uniform. This counters classes that the blend almost never predicts, which hurts balanced accuracy.

### 3.2 Branch A: multi-view TF-IDF + LinearSVC

**Views.** Each passage is rewritten four ways before vectorizing.

| View | Text transform | Example: `The Cat sat, 42 times.` | Analyzer | n-gram range | max features `P_v` | View weight `w_v` |
|---|---|---|---|---|---|---|
| `char` | none | `The Cat sat, 42 times.` | characters | 3-5 | 250,000 | 1.0 |
| `word` | none | tokens `\w+` or single punctuation | words | 1-2 | 150,000 | 0.7 |
| `shape` | each letter run -> `X` (capitalized) or `x` | `X X x, 42 x.` | characters | 2-6 | 60,000 | 0.5 |
| `func` | keep function words and punctuation, others -> `_` | `the _ _ , _ _ .` | whitespace tokens | 1-3 | 50,000 | 0.5 |

All vectorizers use `lowercase=False`, `min_df=2`, `sublinear_tf=True`, and the default L2 row normalization. They are fit on an unlabeled Open Pool sample (up to 5,000 passages per label since v8; 2,500 before), never on validation prompts.

**Layers.**

| # | Layer | Operation | Input | Output | Activation |
|---|---|---|---|---|---|
| A1 | TF-IDF (per view `v`) | `x_v[n,p] = w_v * l2norm_p( (1 + log tf[n,p]) * idf[p] )` | `N` strings | `[N, P_v]` sparse | none (log-scaled term frequency) |
| A2 | Concatenate views | `X[n,p] = hstack_v x_v[n,p]` | 4 x `[N, P_v]` | `[N, P]`, `P = sum_v P_v <= 510,000` | none |
| A2' | Stylometric block (v5, back in v8) | `z[n,j] = clip((log1p(f[n,j]) - mu[j]) / sd[j], -4, 4) * 0.2 / sqrt(27)`; `mu`, `sd` from the `train` rows of the fold; `X <- hstack(X, z)` | 27 stats per passage | `[N, P + 27]` | none |
| A3 | LinearSVC, one-vs-rest | `S = einsum('np,kp->nk', X, W) + b[k]` | `[N, P]` | `[N, 18]` | none (linear margin scores) |
| A4 | Temperature calibration | `LP_A[n,k] = log_softmax_k( S[n,k] / T_A )` | `[N, 18]` | `[N, 18]` | log-softmax |

- Parameters: `W` is `[18, P]`, `b` is `[18]`, so up to about 9.2M weights.
- Row L2 norms: the four TF-IDF views give `sqrt(1.0^2 + 0.7^2 + 0.5^2 + 0.5^2) ~ 1.4`. A standardized 27-dim block has an expected L2 norm of `sqrt(27) ~ 5.2`, so it is scaled by `alpha / sqrt(27)` to an L2 norm of about `alpha = 0.2`. A larger block would be cheaper for the L2-penalized SVC to use and would crowd out the TF-IDF columns.
- Training objective: 18 independent binary problems, squared hinge loss with L2 penalty, `C = 0.1`. `train` rows get weight `w_train = 12` (it won against `{1, 4}` by out-of-fold balanced accuracy in every run since v3, so only 12 is run now). The Open Pool sample (88,107 rows in v8; 45,000 before) is about 40 times larger than the `train` fold (about 2,160 rows), so with equal weights the pool dominates the decision boundary.
- Pool row weights (v8): a pool row of label `L` gets `(N_pool / 18) / N_pool[L]`, so every label has the same total pool weight and the mean pool weight is 1. Two labels have fewer than 5,000 pool passages (`gemini-3.7-flash` 3,541, `qwen3.8-max` 4,566) and get weights 1.38 and 1.07; the other 16 get 0.98.
- `T_A` is picked from `{0.05, ..., 3}` by minimizing out-of-fold negative log-likelihood.

### 3.3 Branch B: Qwen3-4B-Base trunk + LoRA + multi-depth pooling head

Base model config (Qwen3-4B-Base): 36 decoder layers, `D = 2560`, 32 query heads, 8 KV heads, head dim 128, MLP size 9728, RMSNorm `eps = 1e-6`, RoPE `theta = 1,000,000`, no attention bias.
Changes made in this notebook:

- Only the first `keep_layers = 18` decoder blocks are loaded. Blocks 19-36 and the LM head are dropped (`AutoModel`, not `AutoModelForCausalLM`).
- The final RMSNorm is replaced by `nn.Identity()`, so tapped hidden states are the raw residual stream.
- All original weights are frozen. LoRA adapters are added to `q_proj`, `k_proj`, `v_proj`, `o_proj` in every kept block.
- Trunk weights are stored in fp16 on T4 (bf16 on Ampere or newer) and run under autocast. LoRA, LayerNorm, and head weights are fp32. Pooling is done in fp32.

#### B0. Tokenization

| # | Step | Operation | Input | Output |
|---|---|---|---|---|
| B0 | Tokenize | Qwen3 BPE tokenizer, `add_special_tokens=False`, truncate to 384 tokens. **v4 echo**: `ids = x + tok("\n\n") + x`, so `T <= 769` | `B` strings | `ids [B, T]` int64 |
| | Pad + mask | right padding with `pad_token` to the longest sequence in the batch. `pool [B, T]` is 1 only on the second copy (equal to `mask` without echo) | | `ids [B, T]`, `mask [B, T]`, `pool [B, T]` in {0, 1} |

Training batches are length-bucketed (sorted inside chunks of 50 batches, then shuffled) to reduce padding.

#### B1. Embedding

| # | Layer | Operation | Input | Output | Activation |
|---|---|---|---|---|---|
| B1 | Token embedding (frozen) | `h0[b,t,d] = Emb[ids[b,t], d]`, `Emb` is `[V, D]` | `[B, T]` | `[B, T, 2560]` | none |

There is no absolute position embedding. Position enters only through RoPE inside attention.

#### B2. Decoder block, repeated for depth 1..18 (pre-norm)

Input `x = h_{depth-1}` with shape `[B, T, 2560]`. Every projection marked "+ LoRA" is `W x + 2.0 * B(A(x))` (see B3).

```
x [B,T,2560] -----------------------------------+
   |                                            |
 RMSNorm (input_layernorm)                      |
   |                                            |
 q_proj+LoRA   k_proj+LoRA   v_proj+LoRA        |
 [B,T,32,128]  [B,T,8,128]   [B,T,8,128]        |
   |             |             |                |
 q_norm        k_norm          |                |
 RoPE          RoPE            |                |
   +------+------+             |                |
          |                    |                |
 GQA scores [B,8,4,T,T]        |                |
 causal mask, softmax over s   |                |
          +---------+----------+                |
                    |                           |
           weighted sum [B,T,32,128]            |
                    |                           |
           o_proj+LoRA [B,T,2560]               |
                    |                           |
                   (+) <------------------------+
                    |
                    +---------------------------+
                    |                           |
          RMSNorm (post_attention_layernorm)    |
                    |                           |
     gate_proj [B,T,9728]   up_proj [B,T,9728]  |
          SiLU                  |               |
            +------- (*) -------+               |
                      |                         |
             down_proj [B,T,2560]               |
                      |                         |

```

| # | Layer | Operation (einsum) | Input | Output | Activation |
|---|---|---|---|---|---|
| a | RMSNorm (`input_layernorm`) | `u = x * rsqrt(mean_d(x^2) + 1e-6) * gamma[d]` | `[B,T,2560]` | `[B,T,2560]` | none |
| b | Query projection + LoRA | `q = einsum('btd,dhe->bthe', u, Wq)`, `Wq` is `[2560, 32, 128]` | `[B,T,2560]` | `[B,T,32,128]` | none |
| c | Key projection + LoRA | `k = einsum('btd,dge->btge', u, Wk)`, `Wk` is `[2560, 8, 128]` | `[B,T,2560]` | `[B,T,8,128]` | none |
| d | Value projection + LoRA | `v = einsum('btd,dge->btge', u, Wv)`, `Wv` is `[2560, 8, 128]` | `[B,T,2560]` | `[B,T,8,128]` | none |
| e | QK-norm (`q_norm`, `k_norm`) | RMSNorm over `e` for each head, `gamma[e]` of size 128 | `[B,T,32,128]`, `[B,T,8,128]` | same | none |
| f | RoPE | rotate each pair `(e_2i, e_2i+1)` of `q` and `k` by angle `t * theta^(-2i/128)` | same | same | none (rotation) |
| g | Attention scores (GQA) | reshape `q` to `[B,T,8,4,128]`; `a = einsum('btgme,bsge->bgmts', q, k) / sqrt(128)`; set `a = -inf` where `s > t` | `[B,T,8,4,128]`, `[B,T,8,128]` | `[B,8,4,T,T]` | none |
| h | Attention weights | `p = softmax_s(a)` | `[B,8,4,T,T]` | `[B,8,4,T,T]` | **softmax** |
| i | Weighted sum of values | `o = einsum('bgmts,bsge->btgme', p, v)`, reshape to `[B,T,32,128]` | `[B,8,4,T,T]`, `[B,T,8,128]` | `[B,T,32,128]` | none |
| j | Output projection + LoRA | `y = einsum('bthe,hed->btd', o, Wo)`, `Wo` is `[32, 128, 2560]` | `[B,T,32,128]` | `[B,T,2560]` | none |
| k | Residual add | `x2 = x + y` | `[B,T,2560]` x 2 | `[B,T,2560]` | none |
| l | RMSNorm (`post_attention_layernorm`) | `u2 = x2 * rsqrt(mean_d(x2^2) + 1e-6) * gamma2[d]` | `[B,T,2560]` | `[B,T,2560]` | none |
| m | Gate and up projections | `g1 = einsum('btd,df->btf', u2, Wgate)`, `g2 = einsum('btd,df->btf', u2, Wup)` | `[B,T,2560]` | `[B,T,9728]` x 2 | none |
| n | Gated activation (SwiGLU) | `z = SiLU(g1) * g2`, with `SiLU(x) = x * sigmoid(x)` | `[B,T,9728]` x 2 | `[B,T,9728]` | **SiLU** |
| o | Down projection | `y2 = einsum('btf,fd->btd', z, Wdown)` | `[B,T,9728]` | `[B,T,2560]` | none |
| p | Residual add | `h_depth = x2 + y2` | `[B,T,2560]` x 2 | `[B,T,2560]` | none |

Notes:
- Attention dropout is 0 in Qwen3. Each KV head `g` is shared by the 4 query heads `m` in its group.
- Padding is on the right and attention is causal, so real tokens never attend to padding. Outputs at padded positions are computed but removed by the pooling mask in B4.
- Frozen parameters per block: about 100.9M (attention 26.2M + MLP 74.7M + norms). Frozen trunk total: 389.0M (embedding) + 18 x 100.9M = about 2.21B.

#### B3. LoRA adapter (inside b, c, d, j)

For a frozen linear layer `W` of shape `[D_out, D_in]`:

$$ y = W x + \frac{\alpha}{r} \, B \, A \, \mathrm{dropout}(x), \qquad \frac{\alpha}{r} = \frac{32}{16} = 2.0 $$

| # | Step | Operation (einsum) | Input | Output | Activation |
|---|---|---|---|---|---|
| L1 | Frozen path | `y0 = einsum('bti,oi->bto', x, W)` | `[B,T,D_in]` | `[B,T,D_out]` | none |
| L2 | Dropout | `x' = dropout(x, p=0.05)` (training only) | `[B,T,D_in]` | `[B,T,D_in]` | none |
| L3 | Down-projection `A` | `z = einsum('bti,ri->btr', x', A)`, `A` is `[16, D_in]`, Kaiming-uniform init | `[B,T,D_in]` | `[B,T,16]` | none |
| L4 | Up-projection `B` | `delta = einsum('btr,or->bto', z, B)`, `B` is `[D_out, 16]`, zero init | `[B,T,16]` | `[B,T,D_out]` | none |
| L5 | Merge | `y = y0 + 2.0 * delta` | `[B,T,D_out]` x 2 | `[B,T,D_out]` | none |

There is no nonlinearity between `A` and `B`: the update is a purely linear rank-16 correction. Because `B` starts at zero, the model at step 0 is exactly the pretrained Qwen3.

| `q_proj` | 2560 | 4096 (32 x 128) | 106,496 |
| `k_proj` | 2560 | 1024 (8 x 128) | 57,344 |
| `v_proj` | 2560 | 1024 (8 x 128) | 57,344 |
| `o_proj` | 4096 | 2560 | 106,496 |
| per block | | | 327,680 |
| 18 blocks | | | **5,898,240** |

#### B4. Multi-depth pooling head

`hidden_states[0]` is the embedding output and `hidden_states[i]` is the output of block `i`. The head taps `L = 7` depths `i in {2, 4, 6, 9, 12, 15, 18}`. For each tap `l`:

| # | Layer | Operation (einsum) | Input | Output | Activation |
|---|---|---|---|---|---|
| H1 | Masked mean pooling (fp32) | `pooled_l = einsum('btd,bt->bd', h_l, pool) / max(1, einsum('bt->b', pool))` | `[B,T,2560]`, `[B,T]` | `[B,2560]` | none |
| H2 | LayerNorm (one per tap) | `n_l = (pooled_l - mean_d) / sqrt(var_d + 1e-5) * gamma_l[d] + beta_l[d]` | `[B,2560]` | `[B,2560]` | none |
| H3 | Concatenate taps | `z = concat_l n_l` | 7 x `[B,2560]` | `[B,17920]` | none |
| H4 | Dropout | `z = dropout(z, p=0.1)` (training only) | `[B,17920]` | `[B,17920]` | none |
| H5 | Linear classifier | `logits = einsum('bc,kc->bk', z, Whead) + bhead[k]`, `Whead` is `[18, 17920]` | `[B,17920]` | `[B,18]` | none |
| H6 | Output | train: cross-entropy over `softmax_k(logits)` with label smoothing 0.05; inference: `LP_B = log_softmax_k(logits / T_B)` | `[B,18]` | loss scalar / `[B,18]` | **softmax** (inside the loss) / log-softmax |

The per-tap LayerNorm puts every depth on the same scale before concatenation. Without it, deeper residual streams (which have much larger norms) would dominate the head.

#### B5. Parameter count and training

| Part | Parameters | Trainable |
|---|---|---|
| Embedding + 18 decoder blocks | about 2.21B | no |
| LoRA on q/k/v/o (18 blocks) | 5,898,240 | yes |
| 7 LayerNorms (`gamma`, `beta`) | 35,840 | yes |
| Linear head | 322,578 | yes |
| **Total trainable** | **6,256,658** (about 0.28% of the trunk) | |

| | Stage 1 | Stage 2 |
|---|---|---|
| Data | Open Pool: all of it since v9 (label-balanced loss weights, 86 rows whose source prompt id is a `train` prompt id removed); up to 2,500 per label (`match_weight` sampling) before. Trained once | `train` minus fold-`k` prompts, once per fold `k = 0..4`, each starting from the Stage 1 weights |
| Epochs / LR | 1 / 2e-4 | 2 / 5e-5 |
| Time budget | 3.0 h | 1.0 h |

Shared settings: AdamW (weight decay 0.01), head and LayerNorm LR x5, 5% linear warmup then cosine decay, gradient clipping at 1.0, fp16 `GradScaler` on T4, steps with a non-finite loss are skipped.

### 3.4 Blending and prompt centering

| # | Step | Operation | Input | Output | Activation |
|---|---|---|---|---|---|
| C1 | Weighted blend | `LP = w * LP_B + (1 - w) * LP_A`, `w = 0.6` fixed (`CFG["blend_w"]`; `None` picks the best of `{0, 0.1, ..., 1}` on the 5-fold OOF). `LP_B` on test is the mean of the 5 fold models' logits | `[N,18]` x 2 | `[N,18]` | none |
| C2 | Topic clustering (test only) | TF-IDF without English stop words, then agglomerative clustering (cosine distance, average linkage, threshold 0.9) | `N` strings | cluster id per passage | none |
| C3 | Centering | for each cluster `c` with 2 to 18 members: `LP'[i,k] = LP[i,k] - 0.5 * einsum('jk->k', LP[c]) / n_c` for `i in c`; other passages unchanged | `[N,18]` | `[N,18]` | none |
| C4 | Target class shares | `q = softmax_k(LP')`; `tau[k] = 0.65 * mean_n q[n,k] + 0.35 / 18` | `[N,18]` | `[18]` | softmax |
| C5 | Damped Sinkhorn (test with >= 200 rows only) | `beta = 0`; repeat 10 times: `beta += 0.5 * (log tau - log mean_n softmax_k(LP' + beta))` | `[N,18]` | `[18]` per-class bias | softmax |
| C6 | Prediction | `pred[n] = argmax_k (LP'[n,k] + beta[k])` | `[N,18]` | `[N]` label ids | argmax |

### 3.5 Branch C (v7): DeBERTa-v3-large encoder

```
text
  |  DeBERTa-v3 SentencePiece tokenizer (vocab ~128k), adds [CLS] ... [SEP]
  v
ids [B,T], T <= 512
  |  token embedding (no absolute position added here)
  v
[B,T,1024]
  |  24 x encoder block, all weights trained (no LoRA)
  |    disentangled self-attention: 16 heads x 64, scores = c2c + c2p + p2c over relative-position buckets
  |    FFN 1024 -> 4096 -> 1024
  v
h [B,T,1024]  (last layer)
  |  masked mean over tokens ("cls" option: h[:, 0])
  v
z [B,1024] -> Dropout(0.1) -> Linear(1024 -> 18) -> logits [B,18]
  |
LP_C = log_softmax(logits / T_C)
```

- Bidirectional: every token attends to the whole passage, unlike the causal Qwen trunk (which v4 only works around with echo input).
- Pre-trained with replaced-token detection (ELECTRA-style), a token-level "is this word natural here" signal, rather than next-token prediction. Different architecture and objective -> different errors, which is what the blend needs.
- About 435M parameters, all fine-tuned. The Hugging Face `ContextPooler` is not used; the checkpoint has no trained weights for it.
- Same two-stage schedule as branch B: Stage 1 on the Open Pool (all of it since v10, label-balanced loss weights, same rows as branch B), then Stage 2 per fold from the Stage 1 weights -> 2,699 OOF logits. Test logits come from one extra Stage 2 run on all of `train` (`deb_test="full"`).

## 2. 검증: 반드시 프롬프트 단위로 나눕니다

테스트 프롬프트는 train에 없으므로 `prompt_id`로 묶어 폴드를 나눕니다. 일반 KFold는 낙관적인 것이 아니라 **오히려 크게 낮게** 나옵니다. 정답 지문은 검증 폴드에 있고 같은 프롬프트의 다른 모델 지문만 학습에 남아, 분류기가 같은 주제를 "정답이 아닌 모델"과 연결하기 때문으로 보입니다.

아래 셀은 이 차이를 작은 모델로 직접 확인합니다.

````python
cv_folds = list(StratifiedGroupKFold(CFG["n_folds"], shuffle=True, random_state=CFG["seed"]).split(train, y, groups))
random_folds = list(StratifiedKFold(CFG["n_folds"], shuffle=True, random_state=CFG["seed"]).split(train, y))


def quick_cv(folds):
    X = TfidfVectorizer(analyzer="char_wb", ngram_range=(2, 4), sublinear_tf=True, min_df=2,
                        max_features=100_000, lowercase=False).fit_transform(train.text)
    pred = np.zeros(len(train), dtype=int)
    for a, b in folds:
        pred[b] = LinearSVC(C=0.3).fit(X[a], y[a]).predict(X[b])
    return balanced_accuracy_score(y, pred)


print(f"프롬프트 그룹 KFold BA: {quick_cv(cv_folds):.4f}")
print(f"일반 KFold BA        : {quick_cv(random_folds):.4f}")
print(f"무작위 추측 BA       : {1 / K:.4f}")
````

## 3. n-gram 모델

주제 단어에 끌려가지 않도록, 같은 지문을 4가지로 바꿔 봅니다.

- **char**: 문자 3~5-gram. 문장부호와 철자 습관을 잡습니다.
- **word**: 단어와 문장부호의 1~2-gram.
- **shape**: 단어를 `x`/`X`로 치환한 글자 형태. 주제 정보가 거의 사라집니다.
- **func**: 기능어만 남기고 나머지는 `_`로 치환. 문장 구조만 남습니다.
- **문체 피처 (v5, v8에서 다시 넣음)**: 지문 전체를 세어야 나오는 통계 27개입니다. 문장 길이 평균과 표준편차, 단어 길이, 긴 단어 비율, 어휘 다양도(MATTR), 기능어 비율, 100단어당 구두점(—, ;, :, 따옴표 종류 등)과 축약형·인칭 대명사 수 등입니다. 정의는 `analysis_intra_org/features.py`의 `feats`와 같습니다. n-gram 창(최대 6글자, 3단어)으로는 문장 길이 같은 지문 전체 통계를 볼 수 없어서 따로 넣습니다.

**문체 블록을 붙이는 방법.** 피처마다 `log1p`로 큰 값을 누르고, fold의 train 지문으로 구한 평균과 표준편차로 표준화한 뒤 ±4에서 자릅니다. 그다음 `α / √27`을 곱합니다. 표준화한 27차원 벡터의 L2 노름은 평균 √27 ≈ 5.2이므로, 곱한 뒤의 L2 노름은 약 α입니다. TF-IDF 블록의 L2 노름은 √(1.0² + 0.7² + 0.5² + 0.5²) ≈ 1.4입니다. 문체 블록이 이보다 크면 L2 규제가 걸린 SVC가 27개 열에 먼저 기대게 되므로, α = 0.2(TF-IDF의 약 1/7)로 작게 붙입니다. 표준화 통계를 풀이 아니라 train에서 구하는 이유는 풀이 모델마다 주제가 쏠려 있어서입니다. 코드는 v5와 같습니다. v6·v7은 v4에서 갈라져 나와 이 블록이 없었습니다. 풀 2,500개에서 이 블록의 효과는 n-gram OOF 0.4690 → 0.4756이고, 3분기 블렌드에서는 0.5864 → 0.5886으로 오차 범위 안입니다(로컬 계산).

벡터라이저는 라벨 없이 Open Pool에만 맞춰서, 검증 프롬프트의 정보가 들어가지 않게 했습니다. Open Pool의 `match_weight`는 풀의 문장부호 분포를 실제 지문에 맞추는 가중치입니다. 이 값은 train 전체로 계산됐기 때문에, CV 점수가 약간 높게 나올 수 있습니다.

**풀 표본 크기 (v8).** n-gram용 표본을 라벨당 2,500개에서 **5,000개**로 늘립니다(`CFG["ngram_pool_per_label"]`). 표본은 `match_weight`에 비례해 뽑고, TF-IDF 어휘도 그 표본에 다시 맞춥니다. gemini-3.7-flash(3,541개)와 qwen3.8-max(4,566개)는 풀 전체가 5,000개보다 적어서 표본은 88,107행입니다. LoRA와 DeBERTa의 Stage 1 표본(라벨당 2,500개)은 그대로입니다(다시 학습하지 않음).

근거는 로컬 CPU에서 제출 커널과 같은 fold 분할로 잰 OOF입니다(`analysis_pool/`, 문체 블록 포함, train 가중치 12. 3분기 열은 `analysis_v7/whatif_ngram.py`).

| 라벨당 표본 | 풀 행 | n-gram OOF BA | 2분기 블렌드 (lora 0.6 / ng 0.4) | 3분기 블렌드 (lora 0.6 / deb 0.2 / ng 0.2) |
|---|:-:|:-:|:-:|:-:|
| 2,500 (v5와 같은 n-gram) | 45,000 | 0.4756 | 0.5738 | 0.5886 |
| 3,500 | 63,000 | 0.4923 | 0.5768 | 0.5905 |
| 5,000 | 88,107 | 0.5008 | 0.5890 | 0.5960 |
| **5,000 + 라벨 균형 (v8)** | 88,107 | **0.5012** | **0.5905** | **0.5982** |
| 7,500 | 128,107 | 0.5105 | 0.5894 | 0.5979 |
| 7,500 + 라벨 균형 | 128,107 | 0.5134 | 0.5905 | 0.5942 |

- **이 표는 낙관적일 수 있습니다.** 5,000개는 이 OOF를 보고 고른 값입니다. 2,500개 대비 차이의 95% 구간(프롬프트 단위 bootstrap)은 n-gram +0.0256 [+0.011, +0.041], 2분기 블렌드 +0.0167 [+0.005, +0.027]입니다.
- **DeBERTa가 있는 블렌드에서는 이득이 줄어듭니다.** 3분기 고정 가중치에서는 +0.0096 [+0.002, +0.017]입니다. n-gram이 더한 정보의 일부를 DeBERTa가 이미 갖고 있기 때문으로 보입니다(추정).
- **7,500개를 쓰지 않은 이유.** n-gram 단독은 더 높지만 블렌드에서는 5,000개와 같고, fit 시간과 메모리가 약 1.5배입니다.
- **누수 점검.** 풀에는 train 프롬프트와 주제가 가까운 지문이 있습니다. fold마다 그 fold의 검증 프롬프트와 가까운 풀 행을 빼고 다시 재도(풀 행의 약 1.7%, 넓게 잡으면 약 12%) 2,500 → 5,000의 n-gram 차이는 +0.0252, +0.0278로, 빼지 않았을 때(+0.0252)와 같았습니다. 표본을 늘려 생긴 이득이 주제 누수 때문은 아니라는 뜻입니다.
- **풀이 작은 라벨이 밀립니다.** 라벨 균형 가중치 없이 5,000개를 뽑으면 gemini-3.7-flash의 n-gram recall이 0.420 → 0.260으로 떨어집니다(다른 Gemini 모델의 표본만 늘어나기 때문). 균형 가중치를 주면 0.367까지 돌아옵니다.

**풀 행 가중치 (v8, `CFG["ngram_pool_balance"]`).** 라벨 \(L\)의 풀 행에 (풀 행 수 / 18) / (\(L\)의 풀 행 수)를 줍니다. 모든 라벨의 풀 가중치 합이 같아지고 평균은 1입니다. 5,000개를 채운 16개 라벨은 0.979, qwen3.8-max는 1.072, gemini-3.7-flash는 1.382입니다. 전체 BA는 거의 그대로이고(n-gram 0.5008 → 0.5012, 3분기 블렌드 0.5960 → 0.5982, 차이의 구간은 0을 포함), 한 라벨이 밀리는 것을 줄이려고 켭니다.

**train 행 가중치.** 학습 행의 대부분(88,107개)이 Open Pool이고, 실제 대회 지문인 train은 fold당 약 2,160개입니다. 가중치가 같으면 경계가 풀 분포에 맞춰집니다. 그래서 train 행에만 더 큰 가중치를 줍니다. 후보 `CFG["ngram_train_weights"]`를 모두 5-fold로 돌리고 OOF 점수가 가장 높은 값을 쓰는데, v3부터 매번 12가 뽑혀서 12만 둡니다. 풀이 커지면 train의 가중치 몫이 0.42에서 0.27로 줄지만, 몫을 맞추려고 가중치를 23.5로 올린 실행은 나아지지 않았습니다(n-gram 0.5008 → 0.4979). Open Pool이 없으면 모든 행이 train이라 가중치가 의미가 없어서 1.0만 씁니다.

**메모리.** 라벨당 5,000개에서 풀 행렬의 0이 아닌 원소는 약 1.9억 개입니다. `fit_svc`는 fold마다 [train 행 + 풀 행]에 문체 블록을 붙인 행렬을 새로 조립하므로, 조립에 쓴 사본과 float32 행렬을 fit 전에 놓아 줍니다(값은 그대로이고 최대 메모리만 줄어듭니다). 이 셀까지의 프로세스 최대 메모리는 로컬 CPU에서 8.3GB, 제출 커널에서 7.9GB였고, 놓아 주지 않으면 약 11GB입니다(추정). 셀 끝에서 최대 메모리를 출력하고, 풀 행렬은 Qwen과 DeBERTa를 올리기 전에 지웁니다.

**로컬 점검.** 이 셀을 로컬 CPU에서 커널 fold로 그대로 실행하면 표본·행렬·가중치와 5개 fold의 OOF 점수가 풀 크기 실험의 값과 비트 단위로 같습니다(n-gram OOF 0.5012, `analysis_v8/verify_v8.py`, 로그는 `analysis_v8/logs/`).


````python
FUNCTION_WORDS = set("""a an the and or but if while although because as so than that which who whom whose this these
those it its they them their he him his she her we us our you your i me my of in on at by for with from to into over
under about between through during not no nor is are was were be been being has have had do does did will would can
could may might must shall should very more most such only own same just also""".split())


def to_shape(s):
    return re.sub(r"[^\W\d_]+", lambda m: "X" if m[0].isupper() else "x", s)


def to_function(s):
    return " ".join(w.lower() if w.lower() in FUNCTION_WORDS or not w.isalnum() else "_"
                    for w in re.findall(r"\w+|[^\w\s]", s))


VIEWS = [
    ("char", None, dict(analyzer="char", ngram_range=(3, 5), max_features=250_000), 1.0),
    ("word", None, dict(analyzer="word", token_pattern=r"\w+|[^\w\s]", ngram_range=(1, 2), max_features=150_000), 0.7),
    ("shape", to_shape, dict(analyzer="char", ngram_range=(2, 6), max_features=60_000), 0.5),
    ("func", to_function, dict(analyzer="word", token_pattern=r"\S+", ngram_range=(1, 3), max_features=50_000), 0.5),
]


# v5: 문체 피처 27개 (analysis_intra_org/features.py의 feats와 같은 정의). v8에서 v7 위에 그대로 넣음
STYLO_FUNC = set("""a an the and or but if while although because as so than that which who whom whose this these those
it its they them their he him his she her we us our you your i me my of in on at by for with from to into over
under about between through during not no nor is are was were be been being have has had do does did will would
can could should may might must""".split())
SENT_RE = re.compile(r"(?<=[.!?])\s+")
WORD_RE = re.compile(r"[A-Za-z’']+")
CONTR_RE = re.compile(r"\b\w+(?:n['’]t|['’](?:s|re|ll|ve|d|m))\b", re.I)
NOT_BUT_RE = re.compile(r"\bnot\b[^.;:!?]{1,60}?\bbut\b", re.I)
LY_RE = re.compile(r"\b\w{3,}ly\b", re.I)


def stylo_feats(text):
    words = WORD_RE.findall(text)
    nw = max(len(words), 1)
    sents = [s for s in SENT_RE.split(text.strip()) if s]
    sl = np.array([len(WORD_RE.findall(s)) for s in sents]) if sents else np.array([nw])
    lw = [w.lower() for w in words]
    per = lambda c: 100.0 * c / nw
    win = 40
    mattr = (np.mean([len(set(lw[i:i + win])) / win for i in range(0, nw - win + 1, 5)])
             if nw > win else len(set(lw)) / nw)
    first_words = {len(WORD_RE.findall(" ".join(sents[:i]))) for i in range(len(sents))}
    caps = sum(1 for i, w in enumerate(words) if w[0].isupper() and i not in first_words and w not in ("I",))
    return [
        nw, len(sents), sl.mean(), sl.std(),                                    # 단어 수, 문장 수, 문장 길이 평균/표준편차
        np.mean([len(w) for w in words]) if words else 0,                       # 평균 단어 길이
        np.mean([len(w) >= 9 for w in words]) if words else 0,                  # 9글자 이상 단어 비율
        mattr,                                                                  # 어휘 다양도(MATTR, 창 40단어)
        np.mean([w in STYLO_FUNC for w in lw]) if lw else 0,                    # 기능어 비율
        per(text.count("—")), per(text.count(" — ")), per(text.count("–")),     # 이하 100단어당 개수
        per(text.count(";")), per(text.count(":")), per(text.count(",")),
        per(text.count("(")), per(text.count("?")),
        per(text.count("’")), per(text.count("'")),
        per(text.count("“") + text.count("”")), per(text.count('"')),
        per(len(CONTR_RE.findall(text))),                                       # 축약형
        per(sum(w in ("i", "me", "my", "we", "us", "our") for w in lw)),        # 1인칭
        per(sum(w in ("you", "your") for w in lw)),                             # 2인칭
        per(len(NOT_BUT_RE.findall(text))),                                     # not ... but
        per(len(LY_RE.findall(text))),                                          # -ly 단어
        np.mean([bool(re.match(r"(And|But|Yet|So)\b", s)) for s in sents]) if sents else 0,
        caps / nw,                                                              # 문장 첫 단어가 아닌 대문자 단어 비율
    ]


def stylo_raw(texts):
    """지문마다 27개 피처 -> log1p. 표준화는 fold마다 fit_svc에서 합니다."""
    return np.log1p(np.clip(np.array([stylo_feats(t) for t in texts], dtype=np.float64), 0, None)).astype(np.float32)


def stylo_block(S, stats):
    mu, sd = stats
    Z = np.clip((S - mu) / sd, -4, 4) * (CFG["stylo_alpha"] / np.sqrt(S.shape[1]))
    return sparse.csr_matrix(Z.astype(np.float32))


def with_stylo(X, S, stats):
    return X if stats is None else sparse.hstack([X, stylo_block(S, stats)]).tocsr()


def sample_pool(n_per_label, seed=0):
    """라벨당 최대 n개를 match_weight에 비례해 비복원 추출합니다."""
    if pool is None:
        return None
    rng = np.random.default_rng(seed)
    picks = []
    for _, g in pool.groupby("model"):
        w = g.match_weight.fillna(0).to_numpy(dtype=float)
        k = min(n_per_label, int((w > 0).sum()))
        picks.append(g.index.to_numpy()[rng.choice(len(g), k, replace=False, p=w / w.sum())])
    return pool.loc[np.concatenate(picks)].reset_index(drop=True)


class MultiViewVectorizer:
    def __init__(self):
        self.parts = []

    @staticmethod
    def _prep(fn, texts):
        return [fn(t) for t in texts] if fn else list(texts)

    def fit_transform(self, texts):
        out = []
        for _, fn, opts, weight in VIEWS:
            v = TfidfVectorizer(lowercase=False, sublinear_tf=True, min_df=2, dtype=np.float32, **opts)
            out.append(v.fit_transform(self._prep(fn, texts)) * weight)
            self.parts.append((v, fn, weight))
        return sparse.hstack(out).tocsr()

    def transform(self, texts):
        return sparse.hstack([v.transform(self._prep(fn, texts)) * w for v, fn, w in self.parts]).tocsr()


ng_pool = sample_pool(CFG["ngram_pool_per_label"])
vectorizer = MultiViewVectorizer()
if ng_pool is not None:
    X_pool = vectorizer.fit_transform(ng_pool.text)
    y_pool = ng_pool.model.map(L2I).values
    X_train = vectorizer.transform(train.text)
else:
    X_train = vectorizer.fit_transform(train.text)
    X_pool, y_pool = None, None
X_test = vectorizer.transform(test.text)
print("특성 수:", X_train.shape[1], "|", elapsed())

use_stylo = CFG["stylo_alpha"] > 0
S_train = stylo_raw(train.text) if use_stylo else None
S_test = stylo_raw(test.text) if use_stylo else None
S_pool = stylo_raw(ng_pool.text) if use_stylo and ng_pool is not None else None
if use_stylo:
    print(f"문체 피처 {S_train.shape[1]}개, alpha={CFG['stylo_alpha']} | {elapsed()}")

# v8: 풀 행 가중치. 라벨당 5,000개를 뽑으면 풀이 그보다 작은 라벨(gemini-3.7-flash, qwen3.8-max)은 표본이 적어서
# 라벨마다 풀 가중치 합이 달라집니다. 라벨 L의 풀 행에 (풀 행 수 / 라벨 수) / (L의 풀 행 수)를 줘서 합을 같게 맞춥니다.
# 평균이 1이라 train 행 가중치(12)와의 비율은 그대로입니다.
w_pool = None
if X_pool is not None:
    pool_counts = np.bincount(y_pool, minlength=K)
    if CFG["ngram_pool_balance"]:
        w_pool = (len(y_pool) / (pool_counts > 0).sum() / np.maximum(pool_counts, 1))[y_pool]
    else:
        w_pool = np.ones(len(y_pool))
    short = {LABELS[k]: int(c) for k, c in enumerate(pool_counts) if c < CFG["ngram_pool_per_label"]}
    print(f"풀 표본 {len(y_pool):,}행 (라벨당 최대 {CFG['ngram_pool_per_label']:,}개) | 라벨별 {pool_counts.tolist()}")
    print(f"  {CFG['ngram_pool_per_label']:,}개에 못 미치는 라벨: {short if short else '없음'}")
    print(f"풀 행 가중치 (라벨 균형 {'켬' if CFG['ngram_pool_balance'] else '끔'}): 최소 {w_pool.min():.4f} / 최대 {w_pool.max():.4f} "
          f"/ 평균 {w_pool.mean():.4f} | 라벨별 가중치 합 최소 {np.bincount(y_pool, weights=w_pool, minlength=K).min():,.1f} "
          f"/ 최대 {np.bincount(y_pool, weights=w_pool, minlength=K).max():,.1f}")


def fit_svc(train_idx, train_weight):
    # 문체 블록의 표준화 통계는 이 fold의 train 지문으로만 구합니다(풀은 모델마다 주제가 쏠려 있음).
    stats = None
    if use_stylo:
        St = S_train[train_idx]
        stats = (St.mean(0), St.std(0) + 1e-6)
    Xs = [with_stylo(X_train[train_idx], S_train[train_idx] if use_stylo else None, stats)]
    ys, ws = [y[train_idx]], [np.full(len(train_idx), train_weight)]
    if X_pool is not None:
        Xs.append(with_stylo(X_pool, S_pool, stats)); ys.append(y_pool); ws.append(w_pool)
    X_fit = sparse.vstack(Xs).tocsr()
    del Xs  # v8: 조립에 쓴 사본(풀 5,000개에서 약 1.6GB)을 fit 전에 놓아 줍니다.
    X_fit = X_fit.astype(np.float64)  # LinearSVC가 안에서 하는 변환을 미리 해서 float32 행렬도 놓아 줍니다(값은 같음).
    model = LinearSVC(C=CFG["svc_c"])
    model.fit(X_fit, np.concatenate(ys), sample_weight=np.concatenate(ws))
    del X_fit
    model.stylo_stats = stats
    return model


def svc_scores(model, X, S):
    return model.decision_function(with_stylo(X, S, model.stylo_stats))


def ngram_oof(train_weight):
    oof = np.zeros((len(train), K))
    for a, b in cv_folds:
        oof[b] = svc_scores(fit_svc(a, train_weight), X_train[b], S_train[b] if use_stylo else None)
    return oof


weight_grid = CFG["ngram_train_weights"] if X_pool is not None else [1.0]
oof_by_weight, ba_by_weight = {}, {}
for w in weight_grid:
    oof_by_weight[w] = ngram_oof(w)
    ba_by_weight[w] = balanced_accuracy_score(y, oof_by_weight[w].argmax(1))
    folds_ba = [balanced_accuracy_score(y[b], oof_by_weight[w][b].argmax(1)) for _, b in cv_folds]
    print(f"train 가중치 {w:>4}: OOF BA {ba_by_weight[w]:.4f} (fold별 {np.round(folds_ba, 3).tolist()}) | {elapsed()}")
ngram_w = max(ba_by_weight, key=ba_by_weight.get)
oof_ng = oof_by_weight[ngram_w]
print(f"선택한 train 가중치: {ngram_w} -> n-gram OOF BA: {ba_by_weight[ngram_w]:.4f}")

ng_test = svc_scores(fit_svc(np.arange(len(train)), ngram_w), X_test, S_test)
del oof_by_weight
del X_pool, ng_pool, S_pool, w_pool  # Qwen과 DeBERTa를 올리기 전에 풀 행렬을 지웁니다.
gc.collect()
if peak_rss_gb() is not None:
    print(f"n-gram 구간까지 프로세스 최대 메모리(RSS): {peak_rss_gb():.1f}GB | {elapsed()}")
````

## 4. LoRA 분류기

공개 토론의 실험에 따르면 작성 스타일 정보는 Qwen3의 마지막 층보다 **앞쪽 층**에 더 많이 남아 있습니다. 그래서 앞쪽 `keep_layers`개 층만 쓰고, 여러 깊이의 은닉 상태를 평균 풀링해 이어 붙인 뒤 선형 분류합니다.

어텐션 투영층에는 LoRA를 답니다. 원래 가중치 \(W\)는 고정하고 저랭크 행렬만 학습합니다.

$$ \text{output} = Wx + \frac{\alpha}{r} B A x $$

\(B\)를 0으로 초기화하므로 학습 시작 시점의 출력은 원래 모델과 같습니다. 외부 패키지 없이 오프라인에서 돌도록 `peft` 없이 직접 구현했습니다.

학습은 두 단계입니다.
1. **Stage 1**: Open Pool로 각 모델의 전반적인 문체를 학습합니다.
2. **Stage 2**: 진짜 train 지문으로 미세 조정해 대회 데이터의 분포에 맞춥니다.

> **검증 상태:** 이 셀들은 작은 모델(Qwen3 0.6B, Mac MPS)로 코드가 끝까지 도는 것만 확인했습니다. 학습 성능과 T4 GPU에서의 시간, 메모리는 아직 확인하지 못했으므로 첫 실행 때 손실과 홀드아웃 점수를 확인하세요.

첫 번째 폴드의 프롬프트는 학습에서 빼고 홀드아웃으로 씁니다. 각 단계에는 시간 예산이 있어서, 예산이 다 되면 학습을 끊고 다음으로 넘어갑니다.

### v9: Stage 1을 Open Pool 전체로

v8까지 Stage 1은 라벨당 2,500개를 `match_weight`에 비례해 뽑은 45,000개(풀의 19%)로 1 epoch만 학습했습니다. 이 크기는 실험으로 고른 값이 아니라 v2에서 공개 노트북의 기본값을 가져온 것이고, 그 뒤 바꿔 본 적이 없습니다. n-gram은 같은 풀에서 표본을 2,500개에서 5,000개로 늘리자 OOF BA가 0.4693에서 0.5012로 올랐습니다(v8). 데이터셋 설명에는 제작자가 DeBERTa-v3-large를 풀 전체(가중치 없이 약 193,000개)로 학습해 모델 하나로 Public LB 0.587을 냈다는 기록이 있습니다(우리 v7의 3분기 앙상블은 0.556).

| | v8까지 | v9 |
|---|---|---|
| Stage 1 데이터 | 라벨당 2,500개, `match_weight` 비례 비복원 추출 (45,000개) | **풀 전체** (`lora_pool_mode="full"`). 출처의 프롬프트 id가 train `prompt_id`와 같은 86행은 뺌 |
| 라벨별 개수 | 모두 2,500 | 3,541(gemini-3.7-flash) ~ 23,542(gemini-3.8-flash) |
| Stage 1 손실 | cross-entropy (label smoothing 0.05) | 같은 손실에 **라벨 균형 가중치**를 곱함 |
| 체크포인트 | 없음 | `stage1_ckpt_min`분마다. 이어 하기 가능 |

- **라벨 균형 가중치.** 라벨 \(L\)의 풀 행에 \(w_L = (N / 18) / N_L\)를 줍니다(\(N\)은 풀 행 수, \(N_L\)은 라벨 \(L\)의 행 수). 모든 라벨의 가중치 합이 \(N / 18\)로 같아지고 행 평균은 1입니다. 마이크로 배치의 손실은 \(\frac{1}{B}\sum_i w_{y_i}\,\mathrm{CE}_i\)입니다. 가중치 평균이 1이라 가중치 합으로 다시 나누지 않습니다. 가장 작은 라벨(gemini-3.7-flash)의 가중치는 약 3.6, 가장 큰 라벨(gemini-3.8-flash)은 약 0.55입니다.
- **`match_weight`를 손실에 쓰지 않는 이유.** 데이터셋 설명에 "추출용 가중치이고, 시험해 보지 않았다면 손실 가중치로 쓰지 말라"고 적혀 있습니다. 라벨마다 Kish 유효 표본 비율 \((\sum w)^2 / (n \sum w^2)\)이 0.31 ~ 0.68이라, 손실에 곱하면 풀 전체를 넣어도 실제로는 절반 정도만 쓰는 셈이기도 합니다. 대회 지문의 표기에 맞추는 일은 train으로 하는 Stage 2에 맡깁니다.
- **프롬프트 id가 같은 86행.** 출처(`bingli283/ntl-gen-pilot-v1`)가 train 프롬프트를 다시 생성한 것으로 보이는 행입니다(`analysis_pool/logs/leak_check.log`, 같은 번호의 train 프롬프트와 가장 가까움). Stage 1이 검증 fold의 프롬프트를 미리 보지 않도록 뺍니다. 주제만 가까운 풀 행(약 7%)은 v8까지의 표본에도 같은 비율로 있었으므로 그대로 둡니다.
- **이어 하기.** Stage 1 체크포인트에는 학습 파라미터, AdamW·스케줄러·GradScaler 상태, 몇 번째 마이크로 배치까지 썼는지가 들어갑니다. 배치 순서는 seed로 정해지므로 같은 순서로 이어 갑니다. 저장은 가중치 갱신 직후에만 하므로 누적 중간에서 끊기지 않습니다. 풀 행 수나 스텝 수가 다르면 이어 하지 않고 멈춥니다.
- **가중치 파일.** `ntl_lora_v9_stage1.pt`, `ntl_lora_v9_fold{k}.pt`, `ntl_lora_v9_full.pt`로 이름을 바꿨습니다. 저장된 가중치의 `lora_pool_mode`가 지금 설정과 다르면 불러오지 않고 멈춥니다. v4 가중치는 이 값이 없어 `"sample"`로 읽히므로 v9 설정에서는 쓸 수 없습니다.
- **메모리.** 토큰 id를 지문마다 int32 배열로 보관하고 1만 개씩 나눠 토크나이즈합니다. 값은 v8과 같습니다.
- **걸린 시간 (실측, Colab T4 fp16).** Stage 1은 28,869스텝에 13.3시간(스텝당 1.56 ~ 1.76초, GPU 메모리 최대 13.3GB), Stage 2는 fold당 540스텝에 약 16분입니다. Colab 세션이 배정 8시간 뒤에 두 번 끊겨서 체크포인트에서 한 번 이어 했고, 마지막 전체 학습 모델은 같은 설정으로 Kaggle T4에서 학습했습니다. L4·A100(bf16) 경로는 배정되지 않아 실행해 보지 못했습니다.

### v4: echo 입력

Qwen은 causal attention이라 \(t\)번째 토큰은 \(1..t\)번째 토큰만 봅니다. 앞부분 토큰의 은닉 상태에는 뒤 문맥이 전혀 없는 셈입니다. 그래서 지문을 두 번 이어 넣고 **두 번째 사본의 토큰만** 평균 풀링합니다. 두 번째 사본의 각 토큰은 첫 사본 전체를 볼 수 있어서, 양방향 인코더처럼 문맥 전체가 반영된 표현이 됩니다 (Springer et al., *Repetition Improves Language Model Embeddings*, ICLR 2025).

```
input_ids : [지문 토큰 x  ...  ] [\n\n] [지문 토큰 x  ...  ]
pool_mask :  0  0  0  ...  0       0      1  1  1  ...  1
```

- 각 사본은 `max_len`(384) 토큰으로 자르므로 전체 길이는 최대 769입니다. 학습과 추론 시간이 약 2배가 됩니다.
- 메모리를 고려해 Colab에서는 마이크로 배치 4개를 2번 누적해 한 번 갱신합니다(`grad_accum`). 실효 배치 8은 v3과 같습니다.
- 저장된 가중치의 `echo_input` 설정이 현재 설정과 다르면 불러오지 않고 멈춥니다. v3 가중치(`ntl-lora-weights-v3`)는 v4에서 쓸 수 없습니다.
- 대회 공개 솔루션에서는 이 방법을 쓴 사례를 찾지 못했습니다. 상위 공개 커널은 지문을 한 번 넣고 끝에 EOS를 붙이거나, 별도의 양방향 인코더를 둡니다.

### v4: Stage 2를 5-fold로

Stage 1은 Open Pool만 쓰고 train을 쓰지 않으므로 fold와 무관합니다. 그래서 **Stage 1은 한 번만** 학습하고, 그 가중치를 저장한 뒤 **fold마다 Stage 1 가중치에서 다시 시작해 Stage 2**를 학습합니다(새 optimizer, 같은 설정).

- 각 fold 모델은 자기 검증 fold를 예측해 **2,699개 전체의 OOF 로짓**을 만듭니다.
- 테스트 로짓은 fold 모델 5개의 **평균**입니다.
- 가중치는 `ntl_lora_stage1.pt`와 `ntl_lora_fold0.pt` ~ `ntl_lora_fold4.pt`로 저장됩니다(v9부터는 이름에 `v9_`가 붙음). fold 파일 5개를 입력으로 붙이면 재학습 없이 추론만 합니다.
- `CFG["lora_test"] = "full"`이면 테스트 로짓만 **train 전체(2,699개)로 Stage 2를 한 번 더 학습한 모델**(`ntl_lora_full.pt`)에서 가져옵니다. Stage 2 데이터가 25% 많지만, 이 모델 자체는 검증할 데이터가 없습니다. 온도와 블렌드 가중치는 fold 모델 OOF에서 정한 값을 그대로 씁니다. 두 방식 중 어느 쪽이 나은지는 LB로 비교합니다.
- 추가 비용은 Stage 2 네 번(L4 기준 약 3분씩, echo면 약 2배)과 예측입니다.

````python
import torch
from torch import nn
from torch.nn import functional as F


def pick_device():
    if torch.cuda.is_available():
        return torch.device("cuda:0")
    if CFG["force_lora"] and torch.backends.mps.is_available():
        return torch.device("mps")
    return torch.device("cpu")


def find_base_model():
    if os.environ.get("NTL_BASE_MODEL"):
        return os.environ["NTL_BASE_MODEL"]
    found = []
    for cfg_path in glob.glob(str(INPUT_ROOT / "**/config.json"), recursive=True):
        try:
            if json.load(open(cfg_path)).get("model_type") == "qwen3":
                found.append(str(Path(cfg_path).parent))
        except (OSError, ValueError):
            continue
    for wanted in CFG["base_preference"]:
        for path in found:
            if wanted in path:
                return path
    return found[0] if found else None


class LoRALinear(nn.Module):
    def __init__(self, base, r, alpha, dropout):
        super().__init__()
        self.base = base
        self.A = nn.Linear(base.in_features, r, bias=False)
        self.B = nn.Linear(r, base.out_features, bias=False)
        nn.init.kaiming_uniform_(self.A.weight, a=math.sqrt(5))
        nn.init.zeros_(self.B.weight)
        self.scale = alpha / r
        self.drop = nn.Dropout(dropout)

    def forward(self, x):
        delta = self.B(self.A(self.drop(x.to(self.A.weight.dtype))))
        return self.base(x) + (delta * self.scale).to(x.dtype)


def add_lora(trunk, targets=("q_proj", "k_proj", "v_proj", "o_proj")):
    for p in trunk.parameters():
        p.requires_grad_(False)
    n = 0
    for module in list(trunk.modules()):
        for name, child in list(module.named_children()):
            if name in targets and isinstance(child, nn.Linear):
                setattr(module, name, LoRALinear(child, CFG["lora_r"], CFG["lora_alpha"], CFG["lora_dropout"]))
                n += 1
    assert n > 0, "LoRA를 붙일 층을 찾지 못했습니다."
    return n


class StyleClassifier(nn.Module):
    def __init__(self, trunk, depths, n_classes, amp_dtype):
        super().__init__()
        self.trunk, self.depths, self.amp_dtype = trunk, depths, amp_dtype
        hidden = trunk.config.hidden_size
        self.norms = nn.ModuleList(nn.LayerNorm(hidden) for _ in depths)
        self.drop = nn.Dropout(0.1)
        self.head = nn.Linear(hidden * len(depths), n_classes)

    def forward(self, input_ids, attention_mask, pool_mask):
        with torch.autocast("cuda", dtype=self.amp_dtype, enabled=self.amp_dtype is not None):
            out = self.trunk(input_ids=input_ids, attention_mask=attention_mask,
                             output_hidden_states=True, use_cache=False)
        mask = pool_mask.unsqueeze(-1).float()  # echo면 두 번째 사본만, 아니면 attention_mask와 같음
        feats = []
        for depth, norm in zip(self.depths, self.norms):
            pooled = (out.hidden_states[depth].float() * mask).sum(1) / mask.sum(1).clamp(min=1)
            feats.append(norm(pooled))
        return self.head(self.drop(torch.cat(feats, dim=-1)))


def build_classifier(base_dir, device):
    import transformers
    from transformers import AutoConfig, AutoModel, AutoTokenizer
    transformers.logging.set_verbosity_error()  # 잘라낸 층에 대한 로드 경고를 숨깁니다.
    tok = AutoTokenizer.from_pretrained(base_dir, padding_side="right")
    if tok.pad_token is None:
        tok.pad_token = tok.eos_token
    mc = AutoConfig.from_pretrained(base_dir)
    keep = min(CFG["keep_layers"], mc.num_hidden_layers)
    mc.num_hidden_layers, mc.use_cache = keep, False
    if getattr(mc, "layer_types", None) is not None:
        mc.layer_types = mc.layer_types[:keep]
    if device.type == "cuda":
        # T4(compute capability 7.5)는 bf16 연산이 에뮬레이션이라 느리므로 fp16을 씁니다.
        weights = torch.bfloat16 if torch.cuda.get_device_capability(0)[0] >= 8 else torch.float16
        amp = weights
    else:
        weights, amp = torch.float32, None
    trunk = AutoModel.from_pretrained(base_dir, config=mc, dtype=weights)
    trunk.norm = nn.Identity()  # 마지막 정규화는 쓰지 않고 은닉 상태를 그대로 사용
    n_lora = add_lora(trunk)
    depths = [d for d in CFG["depths"] if d <= keep]
    model = StyleClassifier(trunk, depths, K, amp).to(device)
    trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"LoRA 층 {n_lora}개, 학습 파라미터 {trainable:,}개, 깊이 {depths}")
    return model, tok


def make_batches(lengths, batch_size, shuffle, seed=0):
    order = np.arange(len(lengths))
    if not shuffle:
        order = order[np.argsort(lengths)]
        return [order[i:i + batch_size] for i in range(0, len(order), batch_size)]
    rng = np.random.default_rng(seed)
    rng.shuffle(order)
    # 길이가 비슷한 지문끼리 묶어 패딩 낭비를 줄입니다.
    batches = []
    for i in range(0, len(order), batch_size * 50):
        chunk = order[i:i + batch_size * 50]
        chunk = chunk[np.argsort(lengths[chunk])]
        batches += [chunk[j:j + batch_size] for j in range(0, len(chunk), batch_size)]
    rng.shuffle(batches)
    return batches


ECHO_SEP = "\n\n"


def tokenize_all(tok, texts, chunk=10000):
    """토큰 id 배열(int32) 목록과, 풀링을 시작할 위치(echo면 두 번째 사본의 시작, 아니면 0)를 돌려줍니다.
    v9: 풀 전체(약 23만 개)를 다루므로 나눠서 토크나이즈하고 int32 배열로 보관합니다(값은 v8과 같음)."""
    texts = list(texts)
    sep = tok(ECHO_SEP, add_special_tokens=False)["input_ids"] if CFG["echo_input"] else []
    out, starts = [], []
    for i in range(0, len(texts), chunk):
        enc = tok(texts[i:i + chunk], truncation=True, max_length=CFG["max_len"], add_special_tokens=False)["input_ids"]
        for ids in enc:
            ids = ids if ids else [tok.pad_token_id]
            if CFG["echo_input"]:
                out.append(np.asarray(ids + sep + ids, dtype=np.int32))
                starts.append(len(ids) + len(sep))
            else:
                out.append(np.asarray(ids, dtype=np.int32))
                starts.append(0)
    return out, np.array(starts, dtype=int)


def collate(ids_list, starts, idx, pad_id, device):
    seqs = [ids_list[i] for i in idx]
    width = max(map(len, seqs))
    ids = torch.full((len(seqs), width), pad_id, dtype=torch.long)
    mask = torch.zeros((len(seqs), width), dtype=torch.long)
    pool = torch.zeros((len(seqs), width), dtype=torch.long)
    for row, (i, s) in enumerate(zip(idx, seqs)):
        ids[row, :len(s)] = torch.from_numpy(s)
        mask[row, :len(s)] = 1
        pool[row, starts[i]:len(s)] = 1
    return ids.to(device), mask.to(device), pool.to(device)


def train_stage(model, tok, device, texts, labels, epochs, lr, budget_h, name, weights=None, ckpt_path=None):
    """weights: 행마다 손실 가중치(None이면 모두 1). ckpt_path: 주면 CFG["stage1_ckpt_min"]분마다 저장하고, 파일이 있으면 이어 합니다(v9)."""
    params = [p for p in model.parameters() if p.requires_grad]
    head_params = list(model.head.parameters()) + list(model.norms.parameters())
    head_ids = {id(p) for p in head_params}
    optim = torch.optim.AdamW([{"params": [p for p in params if id(p) not in head_ids], "lr": lr},
                               {"params": head_params, "lr": lr * CFG["head_lr_mult"]}], weight_decay=0.01)
    ids_list, starts = tokenize_all(tok, texts)
    lengths = np.array([len(s) for s in ids_list])
    accum = CFG["grad_accum"]
    n_steps = epochs * math.ceil(math.ceil(len(texts) / CFG["batch_size"]) / accum)  # 가중치 갱신 횟수
    warmup = max(1, n_steps // 20)
    sched = torch.optim.lr_scheduler.LambdaLR(
        optim, lambda s: min(1, (s + 1) / warmup) * 0.5 * (1 + math.cos(math.pi * min(1, s / n_steps))))
    use_scaler = device.type == "cuda" and model.amp_dtype == torch.float16
    scaler = torch.amp.GradScaler("cuda", enabled=use_scaler)
    labels = torch.tensor(np.array(labels), dtype=torch.long)
    if weights is not None:
        weights = torch.tensor(np.asarray(weights, dtype=np.float32))
        assert len(weights) == len(labels)
    start, step, skipped, running = time.time(), 0, 0, 0.0
    resume_epoch, resume_bi = 0, 0
    use_ckpt = ckpt_path is not None and CFG["stage1_ckpt_min"] > 0
    if use_ckpt and Path(ckpt_path).exists():
        ck = torch.load(ckpt_path, map_location="cpu", weights_only=False)
        assert ck["n_rows"] == len(texts) and ck["n_steps"] == n_steps, (
            f"{ckpt_path}는 행 {ck['n_rows']:,}개 / {ck['n_steps']}스텝 학습의 체크포인트인데 지금은 {len(texts):,}개 / {n_steps}스텝입니다.")
        missing = model.load_state_dict(ck["state"], strict=False)
        assert not missing.unexpected_keys, missing.unexpected_keys
        optim.load_state_dict(ck["optim"])
        sched.load_state_dict(ck["sched"])
        if ck["scaler"] and scaler.is_enabled():  # bf16(L4, A100)에서는 스케일러를 쓰지 않아 상태가 비어 있음. GPU 종류가 바뀌어도 이어 할 수 있게 함
            scaler.load_state_dict(ck["scaler"])
        resume_epoch, resume_bi = ck["epoch"], ck["bi"]
        step, skipped, running = ck["step"], ck["skipped"], ck["running"]
        print(f"[{name}] 체크포인트에서 이어 하기: {step}/{n_steps} 스텝 (epoch {resume_epoch}, 마이크로 배치 {resume_bi}부터), "
              f"loss {running:.3f}", flush=True)
    last_ck = time.time()
    log_every = max(1, n_steps // (50 if use_ckpt else 10))
    model.train()
    optim.zero_grad(set_to_none=True)
    for epoch in range(resume_epoch, epochs):
        batches = make_batches(lengths, CFG["batch_size"], True, seed=CFG["seed"] + epoch)
        for bi, batch in enumerate(batches):
            if epoch == resume_epoch and bi < resume_bi:
                continue
            if time.time() - start > budget_h * 3600:
                print(f"[{name}] 시간 예산 도달, {step}/{n_steps} 스텝에서 중단")
                model.eval()
                return
            ids, mask, pool = collate(ids_list, starts, batch, tok.pad_token_id, device)
            if weights is None:
                loss = F.cross_entropy(model(ids, mask, pool), labels[batch].to(device), label_smoothing=0.05)
            else:  # v9: 가중치 평균이 1이므로 배치 크기로만 나눕니다(가중치 합으로 다시 나누지 않음).
                per_row = F.cross_entropy(model(ids, mask, pool), labels[batch].to(device), label_smoothing=0.05, reduction="none")
                loss = (per_row * weights[batch].to(device)).mean()
            if torch.isfinite(loss):
                scaler.scale(loss / accum).backward()
                running = 0.98 * running + 0.02 * loss.item() if running else loss.item()
            else:
                skipped += 1  # 이 마이크로 배치만 빼고 누적을 이어갑니다.
            if (bi + 1) % accum and bi + 1 < len(batches):
                continue
            scaler.unscale_(optim)
            torch.nn.utils.clip_grad_norm_(params, 1.0)
            scaler.step(optim)
            scaler.update()
            sched.step()
            optim.zero_grad(set_to_none=True)
            step += 1
            if step % log_every == 0:
                print(f"[{name}] {step}/{n_steps} loss {running:.3f} | {(time.time() - start) / 60:.1f}분", flush=True)
            if use_ckpt and time.time() - last_ck > CFG["stage1_ckpt_min"] * 60 and bi + 1 < len(batches):
                # 가중치 갱신 직후에만 저장하므로 누적 중간에서 끊기지 않습니다. 다음 실행은 마이크로 배치 bi + 1부터 시작합니다.
                torch.save({"state": trainable_state(model), "optim": optim.state_dict(), "sched": sched.state_dict(),
                            "scaler": scaler.state_dict(), "epoch": epoch, "bi": bi + 1, "step": step, "skipped": skipped,
                            "running": float(running), "n_rows": len(texts), "n_steps": n_steps}, str(ckpt_path) + ".tmp")
                os.replace(str(ckpt_path) + ".tmp", ckpt_path)
                last_ck = time.time()
                print(f"[{name}] 체크포인트 저장: {step}/{n_steps} 스텝 | {(time.time() - start) / 60:.1f}분", flush=True)
    print(f"[{name}] 완료: {step}스텝, 비정상 손실로 건너뜀 {skipped}회, {(time.time() - start) / 60:.1f}분")
    model.eval()


@torch.no_grad()
def predict_logits(model, tok, device, texts):
    model.eval()
    ids_list, starts = tokenize_all(tok, texts)
    lengths = np.array([len(s) for s in ids_list])
    out = np.zeros((len(texts), K), dtype=np.float32)
    for batch in make_batches(lengths, CFG["batch_size"] * 2, False):
        ids, mask, pool = collate(ids_list, starts, batch, tok.pad_token_id, device)
        out[batch] = model(ids, mask, pool).float().cpu().numpy()
    bad = ~np.isfinite(out).all(1)
    if bad.any():
        print(f"경고: 비정상 로짓 {bad.sum()}개는 0으로 대체합니다(이 지문은 n-gram 쪽 예측을 따릅니다).")
        out[bad] = 0
    return out


def trainable_state(model):
    keep = {n for n, p in model.named_parameters() if p.requires_grad}
    return {k: v.cpu() for k, v in model.state_dict().items() if k in keep}


# v9: v4 가중치(ntl_lora_fold*.pt)와 섞이지 않도록 이름을 바꿈
FOLD_FILES = [f"ntl_lora_v9_fold{k}.pt" for k in range(CFG["n_folds"])]
FULL_FILE = "ntl_lora_v9_full.pt"
STAGE1_FILE = "ntl_lora_v9_stage1.pt"
STAGE1_CKPT = "ntl_lora_v9_stage1_ckpt.pt"
LORA_OOF_FILE = "ntl_lora_v9_oof.npz"


def load_lora(model, path):
    ckpt = torch.load(path, map_location="cpu", weights_only=True)
    saved_echo = ckpt["cfg"].get("echo_input", False)
    assert saved_echo == CFG["echo_input"], (
        f"{path}는 echo_input={saved_echo}로 학습된 가중치인데 지금 설정은 {CFG['echo_input']}입니다.")
    saved_mode = ckpt["cfg"].get("lora_pool_mode", "sample")
    assert saved_mode == CFG["lora_pool_mode"], (
        f"{path}는 lora_pool_mode={saved_mode!r}로 학습된 가중치인데 지금 설정은 {CFG['lora_pool_mode']!r}입니다.")
    missing = model.load_state_dict(ckpt["state"], strict=False)
    assert not missing.unexpected_keys, missing.unexpected_keys


def cv_after_stage1(model, tok, device):
    """Stage 1 가중치에서 시작해 fold마다 Stage 2를 학습하고, OOF 로짓과 fold 평균 테스트 로짓을 돌려줍니다."""
    stage1_state = {k: v.clone() for k, v in trainable_state(model).items()}
    torch.save({"state": stage1_state, "cfg": CFG, "labels": LABELS}, WORK_DIR / STAGE1_FILE)
    s1 = predict_logits(model, tok, device, train.text.values[cv_folds[0][1]])
    print(f"stage1 fold 0 BA: {balanced_accuracy_score(y[cv_folds[0][1]], s1.argmax(1)):.4f} | {elapsed()}")
    oof, tests = np.zeros((len(train), K), dtype=np.float32), []
    for k, (a, b) in enumerate(cv_folds):
        if (WORK_DIR / FOLD_FILES[k]).exists():  # v9: 이어 하기. 이 실행 폴더에 이미 저장된 fold는 다시 학습하지 않음
            print(f"fold {k}: 저장된 가중치 {WORK_DIR / FOLD_FILES[k]}를 사용합니다.")
            load_lora(model, WORK_DIR / FOLD_FILES[k])
        else:
            model.load_state_dict(stage1_state, strict=False)
            train_stage(model, tok, device, train.text.values[a].tolist(), y[a],
                        CFG["stage2_epochs"], CFG["stage2_lr"], CFG["stage2_budget_h"], f"stage2-fold{k}")
            torch.save({"state": trainable_state(model), "cfg": CFG, "labels": LABELS}, WORK_DIR / FOLD_FILES[k])
        oof[b] = predict_logits(model, tok, device, train.text.values[b])
        tests.append(predict_logits(model, tok, device, test.text.values))
        print(f"fold {k}: LoRA BA {balanced_accuracy_score(y[b], oof[b].argmax(1)):.4f} | {elapsed()}")
    return oof, np.mean(tests, axis=0)


def cv_from_saved(model, tok, device, paths):
    """저장된 fold 가중치로 OOF와 테스트 로짓만 계산합니다 (재학습 없음)."""
    oof, tests = np.zeros((len(train), K), dtype=np.float32), []
    for k, ((_, b), p) in enumerate(zip(cv_folds, paths)):
        load_lora(model, p)
        oof[b] = predict_logits(model, tok, device, train.text.values[b])
        tests.append(predict_logits(model, tok, device, test.text.values))
    return oof, np.mean(tests, axis=0)


def full_test_logits(model, tok, device):
    """train 전체로 Stage 2를 학습한 모델의 테스트 로짓. 저장된 FULL_FILE이 있으면 불러오기만 합니다."""
    saved = find_optional(FULL_FILE)
    if saved is None and (WORK_DIR / FULL_FILE).exists():  # v9: 이어 하기. 이 실행 폴더에 이미 저장된 전체 학습 가중치도 다시 학습하지 않음
        saved = WORK_DIR / FULL_FILE
    if saved is not None:
        print("저장된 전체 학습 가중치를 사용합니다:", saved)
        load_lora(model, saved)
    else:
        # v9: fold 가중치를 입력 데이터셋에서 읽은 실행(cv_from_saved)에서는 Stage 1 가중치가 실행 폴더가 아니라 입력에 있음
        stage1_path = WORK_DIR / STAGE1_FILE if (WORK_DIR / STAGE1_FILE).exists() else find_optional(STAGE1_FILE)
        assert stage1_path is not None, f"{STAGE1_FILE}가 없어 전체 학습 모델을 만들 수 없습니다."
        load_lora(model, stage1_path)
        train_stage(model, tok, device, train.text.tolist(), y,
                    CFG["stage2_epochs"], CFG["stage2_lr"], CFG["stage2_budget_h"], "stage2-full")
        torch.save({"state": trainable_state(model), "cfg": CFG, "labels": LABELS}, WORK_DIR / FULL_FILE)
    return predict_logits(model, tok, device, test.text.values)


def lora_stage1_pool():
    """v9: Stage 1 데이터와 행 가중치. lora_pool_mode="full"이면 풀 전체, "sample"이면 v8까지와 같은 추출."""
    if pool is None:
        return None, None
    if CFG["lora_pool_mode"] == "full":
        lp = pool
    else:
        assert CFG["lora_pool_mode"] == "sample", CFG["lora_pool_mode"]
        lp = sample_pool(CFG["lora_pool_per_label"], seed=CFG["seed"])
    if CFG["lora_drop_train_prompt_ids"] and "prompt_group" in lp.columns:
        tail = lp.prompt_group.fillna("").astype(str).str.split(":").str[-1]
        drop = tail.isin(set(train.prompt_id.astype(str)))
        print(f"Stage 1: 출처의 프롬프트 id가 train prompt_id와 같은 풀 행 {int(drop.sum())}개를 뺍니다.")
        lp = lp[~drop.values]
    lp = lp.reset_index(drop=True)
    yl = lp.model.map(L2I).values
    counts = np.bincount(yl, minlength=K)
    w = None
    if CFG["lora_pool_balance"]:
        w = (len(yl) / (counts > 0).sum() / np.maximum(counts, 1))[yl]
        sums = np.bincount(yl, weights=w, minlength=K)
        print(f"Stage 1 행 가중치(라벨 균형): 최소 {w.min():.3f} / 최대 {w.max():.3f} / 평균 {w.mean():.3f} | "
              f"라벨별 가중치 합 {sums.min():,.1f} ~ {sums.max():,.1f}")
    print(f"Stage 1 데이터 ({CFG['lora_pool_mode']}): {len(lp):,}행 | 라벨별 최소 {counts.min():,} / 최대 {counts.max():,} | "
          + ", ".join(f"{LABELS[k]} {c:,}" for k, c in enumerate(counts)))
    return lp, w
````

````python
holdout_idx = cv_folds[0][1]  # v3(fold 0 홀드아웃)과 비교할 때 씁니다.
lora_oof, lora_test = None, None
base_dir = find_base_model()
device = pick_device()
run_lora = CFG["use_lora"] and base_dir is not None and (device.type == "cuda" or CFG["force_lora"])
print(f"device={device}, base={base_dir}, LoRA 실행={run_lora}")

if run_lora:
    model, tok = build_classifier(base_dir, device)
    saved = [find_optional(f) for f in FOLD_FILES]
    if all(p is not None for p in saved):
        print("저장된 fold 가중치를 사용합니다:", [str(p) for p in saved])
        lora_oof, lora_test = cv_from_saved(model, tok, device, saved)
    else:
        stage1_saved = find_optional(STAGE1_FILE)
        if stage1_saved is None and (WORK_DIR / STAGE1_FILE).exists():
            stage1_saved = WORK_DIR / STAGE1_FILE
        if stage1_saved is not None:
            print("저장된 Stage 1 가중치를 사용합니다:", stage1_saved)
            load_lora(model, stage1_saved)
        else:
            lora_pool, lora_w = lora_stage1_pool()
            if lora_pool is not None:
                if device.type == "cuda":
                    torch.cuda.reset_peak_memory_stats()
                train_stage(model, tok, device, lora_pool.text.tolist(), lora_pool.model.map(L2I).values,
                            CFG["stage1_epochs"], CFG["stage1_lr"], CFG["stage1_budget_h"], "stage1",
                            weights=lora_w, ckpt_path=WORK_DIR / STAGE1_CKPT)
                torch.save({"state": trainable_state(model), "cfg": CFG, "labels": LABELS}, WORK_DIR / STAGE1_FILE)
                (WORK_DIR / STAGE1_CKPT).unlink(missing_ok=True)  # Stage 1 가중치를 저장했으므로 체크포인트는 지움
                if device.type == "cuda":
                    print(f"Stage 1 GPU 메모리 최대 사용량: {torch.cuda.max_memory_allocated() / 2**30:.1f}GB")
                del lora_pool, lora_w
                gc.collect()
        lora_oof, lora_test = cv_after_stage1(model, tok, device)
    if CFG["lora_test"] == "full":
        lora_test = full_test_logits(model, tok, device)
        print("테스트 로짓: train 전체로 학습한 모델 1개")
    else:
        print(f"테스트 로짓: fold 모델 {CFG['n_folds']}개 평균")
    print(f"LoRA OOF BA: {balanced_accuracy_score(y, lora_oof.argmax(1)):.4f} "
          f"(fold 0: {balanced_accuracy_score(y[holdout_idx], lora_oof[holdout_idx].argmax(1)):.4f}) | {elapsed()}")
    # v9: 블렌드 가중치를 따로 다시 고를 수 있게 LoRA 결과를 바로 저장합니다.
    fold_of = np.zeros(len(train), dtype=int)
    for k, (_, b) in enumerate(cv_folds):
        fold_of[b] = k
    np.savez(WORK_DIR / LORA_OOF_FILE, oof=lora_oof, test=lora_test, y=y, fold=fold_of,
             ids=train.id.astype(str).values.astype("U"), test_ids=test.id.astype(str).values.astype("U"))
    print("저장:", WORK_DIR / LORA_OOF_FILE)
    del model
    gc.collect()
    if device.type == "cuda":
        torch.cuda.empty_cache()
else:
    print("LoRA를 건너뜁니다. GPU와 Qwen3 입력을 확인하세요. n-gram만으로 제출 파일을 만듭니다.")
````

## 4-2. DeBERTa 분류기 (v7, 세 번째 분기)

Qwen 분기는 디코더라 \(t\)번째 토큰이 \(1..t\)번째 토큰만 봅니다(v4의 echo 입력은 이를 우회하는 장치입니다). **DeBERTa-v3-large**는 양방향 인코더라 모든 토큰이 지문 전체를 봅니다. 사전학습도 다릅니다. Qwen은 다음 토큰 예측, DeBERTa-v3는 바꿔치기된 토큰 찾기(RTD)입니다. 구조와 학습 신호가 다른 모델은 다른 지문에서 틀릴 가능성이 커서 블렌드에 보탬이 됩니다. 공개 상위권 앙상블 다수가 DeBERTa-v3-large를 포함합니다(`name_that_llm_high_scorers_report.md`).

```
지문 → "[CLS] 지문 [SEP]" (최대 512토큰)       input_ids [B, T]
     → 인코더 24층 (hidden 1024, 헤드 16개 × 64, FFN 4096), 전부 학습
     → 마지막 층 은닉 상태                      h [B, T, 1024]
     → 패딩을 뺀 토큰 평균 (deb_pooling="mean")  z [B, 1024]
     → Dropout(0.1) → Linear(1024 → 18)          logits [B, 18]
```

- 파라미터 약 4.35억 개를 모두 학습합니다(LoRA 없음). Hugging Face 기본 분류기의 pooler(`Linear(1024→1024) → GELU`)는 쓰지 않습니다. 공개 체크포인트에 학습된 가중치가 없는 층이기 때문입니다.
- **학습 일정은 LoRA 분기와 같은 2단계입니다.**
  - Stage 1: Open Pool 45,000개(LoRA와 같은 표본), 1 epoch, 본체 lr 2e-5
  - Stage 2: fold마다 Stage 1 가중치에서 다시 시작해 학습용 4개 fold로 2 epoch, 본체 lr 1e-5
  - 분류 헤드 lr은 본체의 20배입니다. warmup 10% + cosine 감소, gradient L2 노름 1.0에서 자르기, label smoothing 0.05를 씁니다.
  - 실효 배치는 마이크로 배치 8 × 누적 2 = 16입니다.
- **정밀도:** compute capability 8 이상(Colab L4, A100)에서는 bf16 autocast, 그보다 낮은 GPU(T4)에서는 fp32로 돕니다. DeBERTa-v3는 fp16에서 값이 넘쳐 NaN이 나는 사례가 알려져 있어 fp16은 쓰지 않습니다.
- **저장 파일** (`WORK_DIR`)
  - `ntl_deb_oof.npz`: 5-fold OOF 로짓 `[2699, 18]`과 지문 id
  - `ntl_deb_full.pt`: train 전체로 Stage 2를 한 번 더 학습한 모델(fp16, 약 0.9GB). 테스트 로짓은 기본으로 이 모델에서 냅니다(`deb_test="full"`). v4에서 full과 folds의 LB 차이가 0.002였으므로 저장 용량이 작은 쪽을 기본으로 했습니다.
  - `ntl_deb_stage1.pt`: Stage 1 가중치(약 0.9GB). 입력에 붙이면 Stage 1을 건너뜁니다.
  - `ntl_deb_base/`: 토크나이저와 config(가중치 없음). Kaggle에서 원본 DeBERTa 없이 모델 구조를 만들 때 씁니다.
- **Kaggle 제출 커널:** `ntl_deb_oof.npz`, `ntl_deb_full.pt`, `ntl_deb_base/`를 데이터셋으로 붙이면 학습 없이 테스트 추론만 합니다. OOF는 저장된 값을 그대로 씁니다. fold 가중치 5개(약 4.5GB)를 싣지 않기 위해서입니다.

### v10: Stage 1을 Open Pool 전체로

v9에서 LoRA 분기의 Stage 1을 풀 전체로 바꾸자 LoRA OOF가 0.5631에서 0.6072로 올랐습니다. DeBERTa 분기에도 같은 변경을 합니다. 풀 제작자의 기록(DeBERTa-v3-large 하나를 풀 전체로 학습해 Public LB 0.587)도 근거입니다.

| | v9까지 | v10 |
|---|---|---|
| Stage 1 데이터 | 라벨당 2,500개, `match_weight` 비례 추출 (45,000개) | **풀 전체** (`deb_pool_mode="full"`, 230,950행). LoRA v9와 같은 행 |
| Stage 1 손실 | cross-entropy (label smoothing 0.05) | 같은 손실에 **라벨 균형 가중치** (LoRA v9와 같은 식) |
| 정밀도 (T4) | fp32 + gradient checkpointing | **fp16 autocast + GradScaler** (`deb_amp="auto"`) |
| 체크포인트 | 없음 | `deb_stage1_ckpt_min`분마다, AdamW 상태 포함. 이어 하기 가능 |
| 검증 손실 | 기록하지 않음 | 학습 중 기록 (`deb_curve.csv`) |
| Stage 2 | 2 epoch, lr 1e-5 | 같음 |

- **fp16.** v7은 DeBERTa-v3의 fp16 NaN 사례를 피하려고 T4에서 fp32로 돌렸고 Stage 1(45,000개)에 123분이 걸렸습니다. 풀 전체는 fp32로 약 10.5시간입니다. 가중치는 fp32로 두고 연산만 fp16 autocast로 하며, GradScaler가 inf/NaN gradient를 낸 갱신을 건너뜁니다. 건너뛴 갱신 수를 로그에 남깁니다.
- **체크포인트.** 가중치는 fp16, AdamW의 1·2차 모멘트는 bf16으로 줄여 약 2.6GB입니다(fp32 그대로면 5.2GB). 이어 할 때 가중치가 fp16으로 한 번 반올림되는데, Stage 2는 v7부터 fp16으로 저장한 Stage 1 가중치에서 시작했으므로 같은 종류의 반올림입니다.
- **검증 손실.** Stage 1은 train을 보지 않으므로 train 2,699개 전체가 Stage 1의 검증 세트입니다. 손실은 label smoothing과 가중치가 없는 평균 cross-entropy(온도 1)입니다. 학습에는 쓰지 않고 기록만 합니다.

> **실행 기록 (2026-11-25, Colab T4):** L4를 배정받지 못해 T4에서 fp32, `deb_grad_ckpt=True`로 학습했습니다. T4(15GB)는 gradient checkpointing 없이는 마이크로 배치 8에서 메모리가 모자랍니다(켜면 최대 8.1GB). Stage 1은 123분(2,813스텝, fold 0 BA 0.4981), Stage 2는 fold당 약 11.5분이 걸렸고 OOF BA는 0.5053(fold별 0.500 / 0.489 / 0.519 / 0.531 / 0.489)입니다. bf16 경로(L4, A100)는 아직 실행해 보지 못했습니다. 로그는 `analysis_v7/logs/`에 있습니다.


````python
# v10: v7 가중치(ntl_deb_*.pt)와 섞이지 않도록 이름을 바꿈
DEB_STAGE1_FILE, DEB_FULL_FILE, DEB_OOF_FILE = "ntl_deb_v10_stage1.pt", "ntl_deb_v10_full.pt", "ntl_deb_v10_oof.npz"
DEB_STAGE1_CKPT, DEB_OOF_PARTIAL, DEB_CURVE_FILE = "ntl_deb_v10_stage1_ckpt.pt", "ntl_deb_v10_oof_partial.npz", "deb_curve.csv"
DEB_FOLD_FILES = [f"ntl_deb_v10_fold{k}.pt" for k in range(CFG["n_folds"])]
DEB_BASE_DIR = "ntl_deb_base"  # 토크나이저와 config만 (가중치 없음)


def has_weights(d):
    d = Path(d)
    return any(d.glob("*.safetensors")) or any(d.glob("pytorch_model*.bin"))


def find_deberta():
    """DeBERTa 폴더를 찾습니다. 사전학습 가중치가 있는 폴더를 먼저, 없으면 구조만 저장한 ntl_deb_base를 씁니다."""
    if os.environ.get("NTL_DEB_MODEL"):
        return os.environ["NTL_DEB_MODEL"]
    found = []
    for cfg_path in glob.glob(str(INPUT_ROOT / "**/config.json"), recursive=True):
        try:
            if json.load(open(cfg_path)).get("model_type") == "deberta-v2":  # DeBERTa-v3도 이 이름을 씁니다.
                found.append(str(Path(cfg_path).parent))
        except (OSError, ValueError):
            continue
    found.sort(key=lambda d: (not has_weights(d), "large" not in d))
    return found[0] if found else None


class DebertaClassifier(nn.Module):
    def __init__(self, backbone, n_classes, pooling, amp_dtype):
        super().__init__()
        self.backbone, self.pooling, self.amp_dtype = backbone, pooling, amp_dtype
        self.drop = nn.Dropout(0.1)
        self.head = nn.Linear(backbone.config.hidden_size, n_classes)

    def forward(self, input_ids, attention_mask):
        with torch.autocast("cuda", dtype=self.amp_dtype, enabled=self.amp_dtype is not None):
            h = self.backbone(input_ids=input_ids, attention_mask=attention_mask).last_hidden_state
        h = h.float()
        if self.pooling == "cls":
            z = h[:, 0]
        else:
            m = attention_mask.unsqueeze(-1).float()
            z = (h * m).sum(1) / m.sum(1).clamp(min=1)
        return self.head(self.drop(z))


def build_deberta(base_dir, device):
    import transformers
    from transformers import AutoConfig, AutoModel, AutoTokenizer
    transformers.logging.set_verbosity_error()
    try:
        tok = AutoTokenizer.from_pretrained(base_dir)
    except Exception as e:
        # 학습(Colab)과 추론(Kaggle)의 transformers 버전이 다르면 저장된 tokenizer_config.json을 못 읽을 수 있습니다
        # (5.x에서 저장한 것을 4.x에서 읽으면 AttributeError). tokenizer.json을 직접 읽으면 토큰 id가 학습 때와 같습니다.
        from transformers import PreTrainedTokenizerFast
        print(f"AutoTokenizer 로드 실패({type(e).__name__}): tokenizer.json을 직접 읽습니다.")
        tok = PreTrainedTokenizerFast(tokenizer_file=str(Path(base_dir) / "tokenizer.json"), pad_token="[PAD]",
                                      cls_token="[CLS]", sep_token="[SEP]", unk_token="[UNK]", mask_token="[MASK]")
    if has_weights(base_dir):
        backbone = AutoModel.from_pretrained(base_dir)
    else:
        backbone = AutoModel.from_config(AutoConfig.from_pretrained(base_dir))  # 학습한 가중치로 곧 덮어씁니다.
    backbone = backbone.float()
    if CFG["deb_grad_ckpt"]:
        backbone.gradient_checkpointing_enable()
    # v10: T4(cc 7.5)에서도 fp16 autocast. 가중치는 fp32 그대로이고 학습에서는 GradScaler를 씁니다. "fp32"면 v9까지와 같음
    amp = None
    if device.type == "cuda" and CFG["deb_amp"] != "fp32":
        assert CFG["deb_amp"] == "auto", CFG["deb_amp"]
        amp = torch.bfloat16 if torch.cuda.get_device_capability(0)[0] >= 8 else torch.float16
    model = DebertaClassifier(backbone, K, CFG["deb_pooling"], amp).to(device)
    n = sum(p.numel() for p in model.parameters())
    print(f"DeBERTa: {base_dir} | 파라미터 {n:,}개 (전부 학습) | 연산 정밀도 {amp or torch.float32} | 풀링 {CFG['deb_pooling']}")
    return model, tok


def deb_tokenize(tok, texts, chunk=10000):
    """v10: 풀 전체(약 23만 개)를 다루므로 나눠서 토크나이즈하고 int32 배열로 보관합니다(값은 v9와 같음)."""
    texts, out = list(texts), []
    for i in range(0, len(texts), chunk):
        out += [np.asarray(s, dtype=np.int32) for s in
                tok(texts[i:i + chunk], truncation=True, max_length=CFG["deb_max_len"])["input_ids"]]
    return out


def deb_collate(ids_list, idx, pad_id, device):
    seqs = [ids_list[i] for i in idx]
    width = max(map(len, seqs))
    ids = torch.full((len(seqs), width), pad_id, dtype=torch.long)
    mask = torch.zeros((len(seqs), width), dtype=torch.long)
    for row, s in enumerate(seqs):
        ids[row, :len(s)] = torch.from_numpy(s)
        mask[row, :len(s)] = 1
    return ids.to(device), mask.to(device)


def deb_val_loss(model, tok, device, val_ids, val_y):
    """검증 손실(label smoothing·가중치 없는 평균 cross-entropy, 온도 1)과 BA. 학습 중 기록용(v10)."""
    logits = deb_predict(model, tok, device, None, ids_list=val_ids).astype(np.float64)
    lp = logits - logits.max(1, keepdims=True)
    lp -= np.log(np.exp(lp).sum(1, keepdims=True))
    return float(-lp[np.arange(len(val_y)), val_y].mean()), float(balanced_accuracy_score(val_y, logits.argmax(1)))


def deb_train_stage(model, tok, device, texts, labels, epochs, lr, budget_h, name, weights=None, ckpt_path=None, val=None,
                    val_points=10):
    """한 단계를 학습합니다. 끝까지 학습하면 True, 시간 예산이나 NTL_STOP_AT에 걸려 멈추면 False를 돌려줍니다.
    v10: weights = 행마다 손실 가중치(None이면 모두 1). ckpt_path를 주면 CFG["deb_stage1_ckpt_min"]분마다(그리고 멈출 때) 저장하고,
    파일이 있으면 이어 합니다. val = (지문, 라벨)이면 val_points번 검증 손실을 재서 WORK_DIR/deb_curve.csv에 씁니다."""
    head_params = list(model.head.parameters())
    head_ids = {id(p) for p in head_params}
    body_params = [p for p in model.parameters() if id(p) not in head_ids]
    optim = torch.optim.AdamW([{"params": body_params, "lr": lr},
                               {"params": head_params, "lr": lr * CFG["deb_head_lr_mult"]}], weight_decay=0.01)
    scaler = torch.amp.GradScaler("cuda", enabled=model.amp_dtype == torch.float16)
    ids_list = deb_tokenize(tok, texts)
    lengths = np.array([len(s) for s in ids_list])
    bs, accum = CFG["deb_batch_size"], CFG["deb_grad_accum"]
    n_steps = epochs * math.ceil(math.ceil(len(texts) / bs) / accum)  # 가중치 갱신 횟수
    warmup = max(1, int(n_steps * CFG["deb_warmup"]))
    sched = torch.optim.lr_scheduler.LambdaLR(
        optim, lambda s: min(1, (s + 1) / warmup) * 0.5 * (1 + math.cos(math.pi * min(1, s / n_steps))))
    labels = torch.tensor(np.asarray(labels), dtype=torch.long)
    if weights is not None:
        weights = torch.tensor(np.asarray(weights, dtype=np.float32))
        assert len(weights) == len(labels)
    val_ids, val_y = (deb_tokenize(tok, val[0]), np.asarray(val[1])) if val is not None else (None, None)
    val_every = max(1, n_steps // max(1, val_points))
    curve = WORK_DIR / DEB_CURVE_FILE
    start, step, skipped, overflow, running = time.time(), 0, 0, 0, 0.0
    resume_epoch, resume_bi = 0, 0
    use_ckpt = ckpt_path is not None and CFG["deb_stage1_ckpt_min"] > 0
    stop_at = float(os.environ.get("NTL_STOP_AT", "inf"))

    def save_ckpt(epoch, bi):
        opt_sd = optim.state_dict()
        opt_sd["state"] = {k: {n: (v.to(torch.bfloat16) if n != "step" and v.is_floating_point() else v).cpu()
                               for n, v in st.items()} for k, st in opt_sd["state"].items()}
        torch.save({"state": deb_state(model), "optim": opt_sd, "sched": sched.state_dict(), "scaler": scaler.state_dict(),
                    "epoch": epoch, "bi": bi, "step": step, "skipped": skipped, "overflow": overflow, "running": float(running),
                    "n_rows": len(texts), "n_steps": n_steps}, str(ckpt_path) + ".tmp")
        os.replace(str(ckpt_path) + ".tmp", ckpt_path)
        del opt_sd
        gc.collect()

    def log_curve(val_loss=None, val_ba=None):
        new = not curve.exists()
        with open(curve, "a") as f:
            if new:
                f.write("stage,step,n_steps,train_loss,val_loss,val_ba,minutes,skipped,overflow\n")
            f.write(f"{name},{step},{n_steps},{running:.5f},{'' if val_loss is None else f'{val_loss:.5f}'},"
                    f"{'' if val_ba is None else f'{val_ba:.5f}'},{(time.time() - start) / 60:.2f},{skipped},{overflow}\n")

    def do_val():
        if val is None:
            log_curve()
            return
        vl, vb = deb_val_loss(model, tok, device, val_ids, val_y)
        model.train()
        log_curve(vl, vb)
        print(f"[{name}] val {step}/{n_steps}: loss {vl:.4f}, BA {vb:.4f} (train loss {running:.3f}) | {(time.time() - start) / 60:.1f}분", flush=True)

    if use_ckpt and Path(ckpt_path).exists():
        ck = torch.load(ckpt_path, map_location="cpu", weights_only=False)
        assert ck["n_rows"] == len(texts) and ck["n_steps"] == n_steps, (
            f"{ckpt_path}는 행 {ck['n_rows']:,}개 / {ck['n_steps']}스텝 학습의 체크포인트인데 지금은 {len(texts):,}개 / {n_steps}스텝입니다.")
        model.load_state_dict(ck["state"])
        optim.load_state_dict(ck["optim"])  # 모멘트는 파라미터 dtype(fp32)으로 되돌아감
        sched.load_state_dict(ck["sched"])
        if ck["scaler"] and scaler.is_enabled():
            scaler.load_state_dict(ck["scaler"])
        resume_epoch, resume_bi = ck["epoch"], ck["bi"]
        step, skipped, overflow, running = ck["step"], ck["skipped"], ck.get("overflow", 0), ck["running"]
        del ck
        gc.collect()
        print(f"[{name}] 체크포인트에서 이어 하기: {step}/{n_steps} 스텝 (epoch {resume_epoch}, 마이크로 배치 {resume_bi}부터), "
              f"loss {running:.3f}", flush=True)
    model.train()
    if step == 0:
        do_val()  # 시작점 (Stage 2는 Stage 1 모델의 검증 손실)
    last_ck = time.time()
    log_every = max(1, n_steps // (50 if use_ckpt else 10))
    optim.zero_grad(set_to_none=True)
    for epoch in range(resume_epoch, epochs):
        batches = make_batches(lengths, bs, True, seed=CFG["seed"] + epoch)
        for bi, batch in enumerate(batches):
            if epoch == resume_epoch and bi < resume_bi:
                continue
            if bi % accum == 0 and (time.time() - start > budget_h * 3600 or time.time() > stop_at):
                # 누적 경계에서만 멈춥니다(gradient가 비어 있을 때). 체크포인트는 마이크로 배치 bi부터 이어 하도록 저장합니다.
                why = "시간 예산" if time.time() - start > budget_h * 3600 else "NTL_STOP_AT"
                if use_ckpt:
                    save_ckpt(epoch, bi)
                    print(f"[{name}] {why} 도달: {step}/{n_steps} 스텝에서 체크포인트를 저장하고 멈춤 | {(time.time() - start) / 60:.1f}분", flush=True)
                else:
                    print(f"[{name}] {why} 도달, {step}/{n_steps} 스텝에서 중단")
                model.eval()
                return False
            ids, mask = deb_collate(ids_list, batch, tok.pad_token_id, device)
            if weights is None:
                loss = F.cross_entropy(model(ids, mask), labels[batch].to(device), label_smoothing=0.05)
            else:  # 가중치 평균이 1이므로 배치 크기로만 나눕니다(LoRA v9와 같음).
                per_row = F.cross_entropy(model(ids, mask), labels[batch].to(device), label_smoothing=0.05, reduction="none")
                loss = (per_row * weights[batch].to(device)).mean()
            if torch.isfinite(loss):
                scaler.scale(loss / accum).backward()
                running = 0.98 * running + 0.02 * loss.item() if running else loss.item()
            else:
                skipped += 1
            if (bi + 1) % accum and bi + 1 < len(batches):
                continue
            scaler.unscale_(optim)
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            scale_before = scaler.get_scale() if scaler.is_enabled() else None
            scaler.step(optim)  # gradient에 inf/NaN이 있으면 갱신을 건너뛰고 scale을 줄임
            scaler.update()
            if scaler.is_enabled() and scaler.get_scale() < scale_before:
                overflow += 1
            sched.step()
            optim.zero_grad(set_to_none=True)
            step += 1
            if step % log_every == 0:
                print(f"[{name}] {step}/{n_steps} loss {running:.3f} | {(time.time() - start) / 60:.1f}분 | 건너뜀 {skipped}, "
                      f"fp16 overflow {overflow}" + (f", scale {scaler.get_scale():.0f}" if scaler.is_enabled() else ""), flush=True)
            if step % val_every == 0 or step == n_steps:
                do_val()
            if use_ckpt and time.time() - last_ck > CFG["deb_stage1_ckpt_min"] * 60 and bi + 1 < len(batches):
                save_ckpt(epoch, bi + 1)  # 가중치 갱신 직후라 누적 중간이 아님. 다음 실행은 마이크로 배치 bi + 1부터
                last_ck = time.time()
                print(f"[{name}] 체크포인트 저장: {step}/{n_steps} 스텝 | {(time.time() - start) / 60:.1f}분", flush=True)
    print(f"[{name}] 완료: {step}스텝, 비정상 손실로 건너뜀 {skipped}회, fp16 overflow로 건너뛴 갱신 {overflow}회, "
          f"{(time.time() - start) / 60:.1f}분", flush=True)
    model.eval()
    return True


@torch.no_grad()
def deb_predict(model, tok, device, texts, ids_list=None):
    model.eval()
    ids_list = deb_tokenize(tok, texts) if ids_list is None else ids_list
    lengths = np.array([len(s) for s in ids_list])
    out = np.zeros((len(ids_list), K), dtype=np.float32)
    for batch in make_batches(lengths, CFG["deb_batch_size"] * 4, False):
        ids, mask = deb_collate(ids_list, batch, tok.pad_token_id, device)
        out[batch] = model(ids, mask).float().cpu().numpy()
    bad = ~np.isfinite(out).all(1)
    if bad.any():
        print(f"경고: DeBERTa 비정상 로짓 {bad.sum()}개는 0으로 대체합니다.")
        out[bad] = 0
    return out


def deb_state(model, half=True):
    return {k: (v.half() if half and v.is_floating_point() else v).detach().cpu().clone()
            for k, v in model.state_dict().items()}


def save_deb(model, name):
    cfg = {k: v for k, v in CFG.items() if k.startswith("deb_")}
    torch.save({"state": deb_state(model), "cfg": cfg, "labels": LABELS}, WORK_DIR / name)


def load_deb(model, path):
    ckpt = torch.load(path, map_location="cpu", weights_only=True)
    assert ckpt["labels"] == LABELS, f"{path}의 라벨 순서가 다릅니다."
    assert ckpt["cfg"]["deb_pooling"] == CFG["deb_pooling"], (
        f"{path}는 deb_pooling={ckpt['cfg']['deb_pooling']}로 학습됐는데 지금 설정은 {CFG['deb_pooling']}입니다.")
    saved_mode = ckpt["cfg"].get("deb_pool_mode", "sample")  # v7 가중치에는 이 값이 없음
    assert saved_mode == CFG["deb_pool_mode"], (
        f"{path}는 deb_pool_mode={saved_mode!r}로 학습된 가중치인데 지금 설정은 {CFG['deb_pool_mode']!r}입니다.")
    model.load_state_dict(ckpt["state"])  # 전부 학습했으므로 strict


def deb_cv(model, tok, device):
    """현재(Stage 1) 가중치에서 fold마다 Stage 2를 학습해 OOF 로짓을 만들고, 설정에 따라 테스트 로짓을 돌려줍니다."""
    stage1 = deb_state(model, half=False)  # CPU에 fp32로 보관 (약 1.7GB)
    oof, tests, done = np.zeros((len(train), K), dtype=np.float32), [], []
    part = WORK_DIR / DEB_OOF_PARTIAL
    if part.exists() and CFG["deb_test"] == "full":  # v10: 이어 하기. 이 실행 폴더에 저장된 fold OOF는 다시 학습하지 않음
        _p = np.load(part)
        assert (_p["ids"] == train.id.astype(str).values).all()
        oof, done = _p["oof"].astype(np.float32), [int(v) for v in _p["done"]]
        print("이어 하기: 이미 끝난 fold", done)
    fold_of = np.zeros(len(train), dtype=int)
    for k, (_, b) in enumerate(cv_folds):
        fold_of[b] = k
    for k, (a, b) in enumerate(cv_folds):
        if k in done:
            print(f"fold {k}: (저장된 OOF) DeBERTa BA {balanced_accuracy_score(y[b], oof[b].argmax(1)):.4f}")
            continue
        model.load_state_dict(stage1)
        deb_train_stage(model, tok, device, train.text.values[a].tolist(), y[a], CFG["deb_stage2_epochs"],
                        CFG["deb_stage2_lr"], CFG["deb_stage2_budget_h"], f"deb-stage2-fold{k}",
                        val=(train.text.values[b].tolist(), y[b]), val_points=CFG["deb_val_points"][1])
        if CFG["deb_save_folds"]:
            save_deb(model, DEB_FOLD_FILES[k])
        oof[b] = deb_predict(model, tok, device, train.text.values[b])
        if CFG["deb_test"] == "folds":
            tests.append(deb_predict(model, tok, device, test.text.values))
        done.append(k)
        np.savez(str(part) + ".tmp.npz", oof=oof, done=np.array(done), ids=train.id.astype(str).values.astype("U"))
        os.replace(str(part) + ".tmp.npz", part)
        print(f"fold {k}: DeBERTa BA {balanced_accuracy_score(y[b], oof[b].argmax(1)):.4f} | {elapsed()}", flush=True)
    np.savez(WORK_DIR / DEB_OOF_FILE, oof=oof, ids=train.id.astype(str).values.astype("U"), y=y, fold=fold_of)
    if CFG["deb_test"] == "folds":
        return oof, np.mean(tests, axis=0)
    if (WORK_DIR / DEB_FULL_FILE).exists():  # v10: 이어 하기
        print("저장된 전체 학습 모델을 사용합니다:", WORK_DIR / DEB_FULL_FILE)
        load_deb(model, WORK_DIR / DEB_FULL_FILE)
    else:
        model.load_state_dict(stage1)
        deb_train_stage(model, tok, device, train.text.tolist(), y, CFG["deb_stage2_epochs"],
                        CFG["deb_stage2_lr"], CFG["deb_stage2_budget_h"], "deb-stage2-full", val_points=CFG["deb_val_points"][1])
        save_deb(model, DEB_FULL_FILE)
    return oof, deb_predict(model, tok, device, test.text.values)


def deb_from_saved(model, tok, device, oof_path):
    """저장된 OOF와 가중치로 테스트 로짓만 계산합니다 (재학습 없음)."""
    z = np.load(oof_path)
    assert (z["ids"] == train.id.astype(str).values).all(), "저장된 DeBERTa OOF의 지문 순서가 train과 다릅니다."
    if CFG["deb_test"] == "full":
        paths = [find_optional(DEB_FULL_FILE)]
    else:
        paths = [find_optional(f) for f in DEB_FOLD_FILES]
    assert all(p is not None for p in paths), (
        f"deb_test={CFG['deb_test']}에 필요한 DeBERTa 가중치가 입력에 없습니다: "
        f"{DEB_FULL_FILE if CFG['deb_test'] == 'full' else DEB_FOLD_FILES}")
    tests = []
    for p in paths:
        load_deb(model, p)
        tests.append(deb_predict(model, tok, device, test.text.values))
    print("저장된 DeBERTa 결과를 사용합니다:", oof_path, [str(p) for p in paths])
    return z["oof"], np.mean(tests, axis=0)


def deb_stage1_pool():
    """v10: DeBERTa Stage 1 데이터와 행 가중치. LoRA의 lora_stage1_pool(v9)과 같은 규칙이고 deb_* 설정을 씁니다."""
    if pool is None:
        return None, None
    if CFG["deb_pool_mode"] == "full":
        dp = pool
    else:
        assert CFG["deb_pool_mode"] == "sample", CFG["deb_pool_mode"]
        dp = sample_pool(CFG["deb_pool_per_label"], seed=CFG["seed"])
    if CFG["deb_drop_train_prompt_ids"] and "prompt_group" in dp.columns:
        tail = dp.prompt_group.fillna("").astype(str).str.split(":").str[-1]
        drop = tail.isin(set(train.prompt_id.astype(str)))
        print(f"DeBERTa Stage 1: 출처의 프롬프트 id가 train prompt_id와 같은 풀 행 {int(drop.sum())}개를 뺍니다.")
        dp = dp[~drop.values]
    dp = dp.reset_index(drop=True)
    yl = dp.model.map(L2I).values
    counts = np.bincount(yl, minlength=K)
    w = None
    if CFG["deb_pool_balance"]:
        w = (len(yl) / (counts > 0).sum() / np.maximum(counts, 1))[yl]
        print(f"DeBERTa Stage 1 행 가중치(라벨 균형): 최소 {w.min():.3f} / 최대 {w.max():.3f} / 평균 {w.mean():.3f}")
    print(f"DeBERTa Stage 1 데이터 ({CFG['deb_pool_mode']}): {len(dp):,}행 | 라벨별 최소 {counts.min():,} / 최대 {counts.max():,}")
    return dp, w
````

````python
deb_oof, deb_test = None, None
deb_dir = find_deberta()
run_deb = CFG["use_deberta"] and deb_dir is not None and (device.type == "cuda" or CFG["force_lora"])
print(f"DeBERTa 폴더={deb_dir}, 실행={run_deb}")

if run_deb:
    if device.type == "cuda":
        torch.cuda.reset_peak_memory_stats()
    model, dtok = build_deberta(deb_dir, device)
    saved_oof = find_optional(DEB_OOF_FILE)
    if saved_oof is not None:
        deb_oof, deb_test = deb_from_saved(model, dtok, device, saved_oof)
    else:
        assert has_weights(deb_dir) or (WORK_DIR / DEB_STAGE1_CKPT).exists() or find_optional(DEB_STAGE1_CKPT) is not None, (
            f"{deb_dir}에 사전학습 가중치가 없어 학습할 수 없습니다.")
        base_out = WORK_DIR / DEB_BASE_DIR
        base_out.mkdir(parents=True, exist_ok=True)
        dtok.save_pretrained(base_out)
        model.backbone.config.save_pretrained(base_out)
        stage1_path = find_optional(DEB_STAGE1_FILE)
        if stage1_path is None and (WORK_DIR / DEB_STAGE1_FILE).exists():
            stage1_path = WORK_DIR / DEB_STAGE1_FILE
        if stage1_path is not None:
            print("저장된 DeBERTa Stage 1 가중치를 사용합니다:", stage1_path)
            load_deb(model, stage1_path)
        else:
            deb_pool, deb_w = deb_stage1_pool()
            if deb_pool is not None:
                ck_in = find_optional(DEB_STAGE1_CKPT)  # Kaggle에서 이어 할 때: 입력 데이터셋의 체크포인트를 실행 폴더로 복사
                if ck_in is not None and not (WORK_DIR / DEB_STAGE1_CKPT).exists():
                    import shutil
                    shutil.copy(ck_in, WORK_DIR / DEB_STAGE1_CKPT)
                finished = deb_train_stage(model, dtok, device, deb_pool.text.tolist(), deb_pool.model.map(L2I).values,
                                           CFG["deb_stage1_epochs"], CFG["deb_stage1_lr"], CFG["deb_stage1_budget_h"], "deb-stage1",
                                           weights=deb_w, ckpt_path=WORK_DIR / DEB_STAGE1_CKPT,
                                           val=(train.text.tolist(), y), val_points=CFG["deb_val_points"][0])
                if not finished and CFG["deb_stage1_ckpt_min"] > 0:
                    raise SystemExit("DeBERTa Stage 1이 끝나지 않았습니다. 체크포인트(" + DEB_STAGE1_CKPT + ")에서 이어 하세요.")
                save_deb(model, DEB_STAGE1_FILE)
                load_deb(model, WORK_DIR / DEB_STAGE1_FILE)  # 이어 하기와 같게: 모든 fold가 저장된(fp16) Stage 1 가중치에서 시작
                (WORK_DIR / DEB_STAGE1_CKPT).unlink(missing_ok=True)
                if device.type == "cuda":
                    print(f"DeBERTa Stage 1 GPU 메모리 최대 사용량: {torch.cuda.max_memory_allocated() / 2**30:.1f}GB")
                del deb_pool, deb_w
                gc.collect()
        s1 = deb_predict(model, dtok, device, train.text.values[holdout_idx])
        print(f"DeBERTa Stage 1 fold 0 BA: {balanced_accuracy_score(y[holdout_idx], s1.argmax(1)):.4f} | {elapsed()}")
        deb_oof, deb_test = deb_cv(model, dtok, device)
    print(f"테스트 로짓: {'train 전체로 학습한 모델 1개' if CFG['deb_test'] == 'full' else 'fold 모델 평균'}")
    print(f"DeBERTa OOF BA: {balanced_accuracy_score(y, deb_oof.argmax(1)):.4f} "
          f"(fold 0: {balanced_accuracy_score(y[holdout_idx], deb_oof[holdout_idx].argmax(1)):.4f}) | {elapsed()}")
    if device.type == "cuda":
        print(f"DeBERTa 구간 GPU 메모리 최대 사용량: {torch.cuda.max_memory_allocated() / 2**30:.1f}GB")
    del model
    gc.collect()
    if device.type == "cuda":
        torch.cuda.empty_cache()
else:
    print("DeBERTa를 건너뜁니다. GPU와 DeBERTa 입력(사전학습 모델 또는 ntl_deb_base + 저장 결과)을 확인하세요.")
````

## 5. 블렌딩

분기마다 점수를 온도 조정한 로그 확률로 바꾼 뒤 가중합합니다. v4부터 LoRA도, v7부터 DeBERTa도 5-fold OOF가 있으므로 온도와 가중치를 모두 같은 2,699개 OOF에서 고릅니다. v3과 비교할 수 있도록 fold 0만의 점수도 함께 출력합니다.

**온도**는 `softmax(점수 / T)`의 `T`입니다. argmax는 바꾸지 않고 확률이 얼마나 뾰족한지만 바꿉니다. n-gram(SVM 마진)과 LoRA·DeBERTa(로짓)는 점수 스케일이 달라서, 분기마다 OOF에서 음의 로그 우도(정답 클래스 log 확률의 평균에 −1을 곱한 값)가 가장 작은 `T`를 골라 같은 기준의 확률로 맞춘 뒤 섞습니다.

**가중치 (v7)**

$$ LP = w_{\text{lora}}\,LP_{\text{lora}} + w_{\text{deb}}\,LP_{\text{deb}} + w_{\text{ng}}\,LP_{\text{ng}}, \qquad w_{\text{ng}} = 1 - w_{\text{lora}} - w_{\text{deb}} $$

- `CFG["blend3"]`가 `None`이면 \((w_{\text{lora}}, w_{\text{deb}})\)를 0.1 간격으로, 합이 1 이하인 66개 조합을 모두 OOF BA로 비교해 가장 높은 조합을 씁니다.
- 같은 OOF로 고르므로 그 점수는 낙관적입니다. 그래서 **nested BA**도 출력합니다. fold 하나를 빼고 나머지 4개 fold에서 가장 좋은 조합을 고른 뒤, 그 조합을 뺀 fold에 적용해 5개 fold를 이어 붙인 점수입니다.
- 첫 실행 결과로 `blend3`을 lora 0.6, deb 0.2, ng 0.2로 고정했습니다(OOF BA 0.5864, nested 0.5823. 같은 OOF에서 v6 가중치 lora 0.6, ng 0.4는 0.5716). v3에서 `w`를 0.6으로 고정한 것과 같은 이유입니다. 다시 고르려면 `None`으로 둡니다.
- **v8:** `blend3`은 v7 값을 그대로 둡니다. n-gram이 강해지면 OOF 최고 조합이 lora 0.4 / deb 0.3 / ng 0.3 쪽으로 옮겨 가지만(로컬 계산 0.5997), 고정값(0.5982)과의 차이가 0.0015(지문 4개)입니다. 셀이 출력하는 상위 5개 조합으로 다시 확인합니다.
- DeBERTa가 없으면 v6과 같이 LoRA 가중치 `blend_w`(0.6)로 두 분기만 섞습니다.
- 분기별 단독 BA와, LoRA와 DeBERTa의 1순위가 갈리는 지문 비율도 출력합니다. 앙상블 이득은 두 모델이 서로 다른 지문을 틀릴 때 생깁니다.


````python
import itertools


def log_softmax_np(x, temperature=1.0):
    x = x / temperature
    x = x - x.max(1, keepdims=True)
    return x - np.log(np.exp(x).sum(1, keepdims=True))


def best_temperature(scores, labels, grid=(0.05, 0.1, 0.2, 0.3, 0.5, 0.75, 1, 1.5, 2, 3)):
    nll = [-log_softmax_np(scores, t)[np.arange(len(labels)), labels].mean() for t in grid]
    return grid[int(np.argmin(nll))]


def ba(lp, idx=None):
    idx = np.arange(len(y)) if idx is None else idx
    return balanced_accuracy_score(y[idx], lp[idx].argmax(1))


def blend(lps, w):
    return sum(w[k] * lps[k] for k in w if w[k])


T_NG = best_temperature(oof_ng, y)
ng_oof_lp, ng_test_lp = log_softmax_np(oof_ng, T_NG), log_softmax_np(ng_test, T_NG)
oof_lps, test_lps = {"ng": ng_oof_lp}, {"ng": ng_test_lp}
print("n-gram 온도:", T_NG)
lora_lp = lora_test_lp = deb_lp = deb_test_lp = None
if lora_oof is not None:
    T_LORA = best_temperature(lora_oof, y)
    lora_lp, lora_test_lp = log_softmax_np(lora_oof, T_LORA), log_softmax_np(lora_test, T_LORA)
    oof_lps["lora"], test_lps["lora"] = lora_lp, lora_test_lp
    print("LoRA 온도:", T_LORA)
if deb_oof is not None:
    T_DEB = best_temperature(deb_oof, y)
    deb_lp, deb_test_lp = log_softmax_np(deb_oof, T_DEB), log_softmax_np(deb_test, T_DEB)
    oof_lps["deb"], test_lps["deb"] = deb_lp, deb_test_lp
    print("DeBERTa 온도:", T_DEB)

for name, lp in oof_lps.items():
    print(f"{name:>5} 단독 OOF BA {ba(lp):.4f} (fold 0: {ba(lp, holdout_idx):.4f})")
if lora_lp is not None and deb_lp is not None:
    pl, pdb = lora_lp.argmax(1), deb_lp.argmax(1)
    print(f"LoRA와 DeBERTa의 1순위가 다른 지문 {(pl != pdb).mean():.1%} | 정확도: LoRA {(pl == y).mean():.1%}, "
          f"DeBERTa {(pdb == y).mean():.1%}, 둘 중 하나라도 맞힘 {((pl == y) | (pdb == y)).mean():.1%}")

others = [n for n in oof_lps if n != "ng"]
grid = []
for combo in itertools.product(range(11), repeat=len(others)):
    if sum(combo) <= 10:
        w = {n: c / 10 for n, c in zip(others, combo)}
        w["ng"] = round(1 - sum(w.values()), 1)
        grid.append(w)
scores = np.array([ba(blend(oof_lps, w)) for w in grid])

if "deb" in oof_lps:
    fixed = CFG["blend3"]
elif "lora" in oof_lps:
    fixed = None if CFG["blend_w"] is None else {"lora": CFG["blend_w"], "ng": 1 - CFG["blend_w"]}
else:
    fixed = {"ng": 1.0}

fmt = lambda w: ", ".join(f"{k}={v:.1f}" for k, v in w.items())
for i in np.argsort(-scores)[:5]:
    print(f"  OOF BA {scores[i]:.4f} | {fmt(grid[i])}")
if "lora" in oof_lps and "deb" in oof_lps:
    ref = {"lora": 0.6, "deb": 0.0, "ng": 0.4}
    print(f"  (v6 가중치 {fmt(ref)}: OOF BA {ba(blend(oof_lps, ref)):.4f})")

if fixed is None:
    weights = grid[int(scores.argmax())]
    nested = np.zeros(len(y), dtype=int)
    for a, b in cv_folds:
        w_in = grid[int(np.argmax([ba(blend(oof_lps, w), a) for w in grid]))]
        nested[b] = blend(oof_lps, w_in)[b].argmax(1)
    print(f"선택한 가중치 (OOF 최고): {fmt(weights)} | OOF BA {scores.max():.4f}, "
          f"nested BA {balanced_accuracy_score(y, nested):.4f}")
else:
    weights = fixed
    print(f"고정한 가중치: {fmt(weights)}")
blend_w = weights.get("lora", 0.0)
eval_idx = np.arange(len(train))
eval_lp, test_lp = blend(oof_lps, weights), blend(test_lps, weights)

np.savez(WORK_DIR / "oof_v10.npz", y=y, ids=train.id.astype(str).values.astype("U"),
         **{f"{k}_logits": v for k, v in {"ng": oof_ng, "lora": lora_oof, "deb": deb_oof}.items() if v is not None})
print(f"최종 검증 BA (OOF, n={len(eval_idx)}): {ba(eval_lp):.4f} | fold 0만 (v3과 비교용): {ba(eval_lp, holdout_idx):.4f}")
````

### 5-1. 회사 게이트 (v6)

v4 오답 1,147개 중 681개(59%)가 같은 회사 모델끼리 헷갈린 것입니다. 정답 회사를 알려 주고 그 안에서만 고르면 v4 OOF BA가 0.575에서 0.694로 오릅니다. 반면 회사를 먼저 고르고 그 안에서만 고르는 딱딱한 2단계 분류는 0.567로 오히려 떨어집니다. GLM은 회사 단위 recall이 0.44로, 회사를 정하는 단계부터 자주 틀리기 때문입니다.

그래서 회사 판단은 블렌드에 맡기고 **회사 안 분포만** 고칩니다.

1. 회사 \(F\)(GPT 5개, Gemini 4개, Claude 5개, GLM 2개)마다, 정답이 \(F\)인 train 지문으로 로지스틱 회귀 \(s_F\)를 학습합니다. 입력은 n-gram과 LoRA 각각의 \(F\) 열 log 확률이고, 출력은 \(F\) 안의 모델입니다.
2. 블렌드 1순위가 \(F\) 안인 지문만 바꿉니다. 회사 확률 질량 \(M_F = \log\sum_{k\in F} e^{\ell_k}\)는 그대로 두고, 회사 안 분포를 비율 \(\alpha_F\)만큼 \(s_F\) 쪽으로 옮깁니다.

$$ \ell^{\text{new}}_{k} = M_F + \log\operatorname{softmax}_{k\in F}\big((1-\alpha_F)(\ell_k - M_F) + \alpha_F \log s_F(k)\big) $$

- OOF 점수는 \(s_F\)도 같은 `cv_folds`로 fold 밖에서 학습해 계산합니다. 테스트용 \(s_F\)는 OOF 전체로 학습합니다.
- \(\alpha_F\)는 v4 OOF에서 골랐습니다. fold 하나를 빼고 나머지 4개로 고르는 nested 방식으로는 +0.0115(프롬프트 bootstrap 95% 구간 +0.003~+0.020)입니다. 이 셀이 출력하는 게이트 OOF BA(+0.013)는 \(\alpha\)를 같은 OOF로 골랐으므로 약간 낙관적입니다.
- 비교 실험(`analysis_v6/`): V40A처럼 모든 지문에 잔차를 더하는 방식은 nested 기준 이득이 거의 없었습니다(+0.0004). Open Pool로 학습한 회사 전용 n-gram SVC를 입력에 더해도 나아지지 않았습니다(+0.0052).

- **v7:** DeBERTa가 있으면 LR 입력에 DeBERTa의 \(F\) 열 log 확률도 넣습니다(입력 차원 = 분기 수 × 회사 모델 수). \(\alpha_F\)는 v4 OOF에서 고른 값을 그대로 쓰고, 셀이 출력하는 \(\alpha\)별 OOF BA로 다시 확인합니다. `CFG["gate_alpha"] = None`이면 회사마다 OOF 최고값을 고릅니다(낙관적).


````python
from scipy.special import log_softmax, logsumexp
from sklearn.linear_model import LogisticRegression

FAMILIES = {p: [L2I[l] for l in LABELS if l.startswith(p)] for p in ("gpt", "gemini", "claude", "glm")}


def gate_features(cols, lps):
    return np.hstack([lp[:, cols] for lp in lps])


def fit_specialist(cols, rows, lps):
    """회사 cols의 정답 지문(rows 중)만으로, 각 분기의 회사 열 log 확률 -> 회사 안 모델을 맞히는 LR."""
    rows = rows[np.isin(y[rows], cols)]
    target = np.searchsorted(cols, y[rows])
    return LogisticRegression(C=CFG["gate_c"], max_iter=3000).fit(gate_features(cols, lps)[rows], target)


def apply_gate(lp, cols, spec_lp, alpha):
    """1순위가 회사 cols 안인 행만, 회사 확률 질량은 그대로 두고 회사 안 분포를 특화 분류기 쪽으로 alpha만큼 옮깁니다."""
    out = lp.copy()
    sub = lp[:, cols]
    mass = logsumexp(sub, axis=1, keepdims=True)
    new = log_softmax((1 - alpha) * (sub - mass) + alpha * spec_lp, axis=1) + mass
    gated = np.isin(lp.argmax(1), cols)[:, None]
    out[:, cols] = np.where(gated, new, sub)
    return out


def within_acc(lp, cols):
    m = np.isin(y, cols)
    return (lp[m][:, cols].argmax(1) == np.searchsorted(cols, y[m])).mean()


eval_lp_pre, test_lp_pre = eval_lp, test_lp
if CFG["use_gate"] and lora_oof is not None:
    gate_oof_lps = [ng_oof_lp, lora_lp] + ([deb_lp] if deb_lp is not None else [])
    gate_test_lps = [ng_test_lp, lora_test_lp] + ([deb_test_lp] if deb_test_lp is not None else [])
    print("게이트 입력 분기:", "n-gram, LoRA" + (", DeBERTa" if deb_lp is not None else ""))
    gated_oof, gated_test = eval_lp.copy(), test_lp.copy()
    for name, cols in FAMILIES.items():
        # OOF: 특화 분류기도 fold 밖에서 학습해 검증 점수가 새지 않게 합니다.
        spec_oof = np.zeros((len(train), len(cols)))
        for a, b in cv_folds:
            m = fit_specialist(cols, a, gate_oof_lps)
            spec_oof[b] = m.predict_log_proba(gate_features(cols, gate_oof_lps)[b])
        spec_test = fit_specialist(cols, np.arange(len(train)), gate_oof_lps).predict_log_proba(
            gate_features(cols, gate_test_lps))
        grid = {a_: balanced_accuracy_score(y, apply_gate(gated_oof, cols, spec_oof, a_).argmax(1))
                for a_ in (0, 0.25, 0.5, 0.75, 1.0)}
        alpha = CFG["gate_alpha"][name] if CFG["gate_alpha"] else max(grid, key=grid.get)
        inC = np.isin(y, cols)
        spec_acc = (spec_oof[inC].argmax(1) == np.searchsorted(cols, y[inC])).mean()
        print(f"[{name}] 회사 안 정확도 블렌드 {within_acc(eval_lp, cols):.3f}, 특화 분류기 {spec_acc:.3f} | "
              f"alpha {alpha} 사용, alpha별 OOF BA {({k: round(float(v), 4) for k, v in grid.items()})}")
        gated_oof = apply_gate(gated_oof, cols, spec_oof, alpha)
        gated_test = apply_gate(gated_test, cols, spec_test, alpha)
    eval_lp, test_lp = gated_oof, gated_test
    print(f"OOF BA: 블렌드 {balanced_accuracy_score(y, eval_lp_pre.argmax(1)):.4f} -> 게이트 "
          f"{balanced_accuracy_score(y, eval_lp.argmax(1)):.4f} | OOF에서 바뀐 지문 "
          f"{(eval_lp.argmax(1) != eval_lp_pre.argmax(1)).sum()}개, 테스트에서 바뀐 지문 "
          f"{(test_lp.argmax(1) != test_lp_pre.argmax(1)).sum()}개 / {len(test)}개")
else:
    print("회사 게이트를 건너뜁니다 (use_gate=False 또는 LoRA 없음).")
````

## 6. 프롬프트 중심화 (테스트 내부 관계 활용)

같은 프롬프트에 답한 지문들은 주제를 공유합니다. 그 묶음의 평균 점수를 빼면 주제 신호가 줄고 문체 차이가 남습니다. 테스트에는 `prompt_id`가 없으므로 주제 단어 TF-IDF로 지문을 묶습니다. 대회 주최 측은 테스트 지문 간 관계를 활용해도 된다고 토론에서 답했습니다.

n-gram OOF로 시뮬레이션한 결과입니다(Open Pool 라벨당 1,000개, 중심화 강도 0.5, 묶음 거리 0.9, 5회 평균).

| 프롬프트당 지문 수 | 중심화 전 | 중심화 후 |
|---|---|---|
| 1 | 0.429 | 0.420 |
| 2 | 0.450 | 0.453 |
| 4 | 0.428 | 0.451 |
| 8 | 0.432 | 0.461 |

프롬프트당 지문이 1개뿐이면 오히려 손해입니다. 실제 테스트 구성은 공개되지 않았고, LoRA 블렌드에서는 측정하지 않았습니다.

묶음 크기가 2~18일 때만 적용합니다. 인터랙티브 실행(테스트 3행)에서는 묶음이 생기지 않아 아무 변화가 없습니다.

````python
def topic_clusters(texts, distance):
    if len(texts) < 2:
        return np.zeros(len(texts), dtype=int)
    X = TfidfVectorizer(stop_words="english", sublinear_tf=True, dtype=np.float32).fit_transform(texts).toarray()
    return AgglomerativeClustering(n_clusters=None, metric="cosine", linkage="average",
                                   distance_threshold=distance).fit_predict(X)


def center_by_cluster(lp, clusters, strength):
    out = lp.copy()
    for c in np.unique(clusters):
        idx = np.where(clusters == c)[0]
        if 2 <= len(idx) <= 18:
            out[idx] = lp[idx] - strength * lp[idx].mean(0)
    return out


final_lp = test_lp
if CFG["use_centering"]:
    clusters = topic_clusters(test.text.tolist(), CFG["cluster_distance"])
    sizes = pd.Series(clusters).value_counts()
    print(f"테스트 묶음 {len(sizes)}개, 중심화 대상 지문 {sizes[(sizes >= 2) & (sizes <= 18)].sum()}개 / {len(test)}개")
    final_lp = center_by_cluster(test_lp, clusters, CFG["center_strength"])
````

## 7. 클래스 비율 보정 (Damped Sinkhorn)

balanced accuracy는 클래스별 recall의 평균이라, 블렌드가 거의 예측하지 않는 클래스가 있으면 그 클래스의 recall이 0에 가까워져 점수가 크게 깎입니다. 그래서 클래스 \(k\)마다 로그 보정값 \(\beta_k\)를 더해 예측 비율을 조정합니다.

목표 비율을 완전 균등(1/18)으로 잡지 않고 현재 예측 비율과 섞습니다. 테스트의 실제 클래스 비율이 정확히 균등하다는 보장이 없기 때문입니다.

$$ \tau_k = (1-\rho)\,\bar q_k + \rho \cdot \tfrac{1}{18}, \qquad \bar q_k = \tfrac{1}{N}\textstyle\sum_n \mathrm{softmax}(LP_n)_k $$

그다음 \(\beta\)를 0에서 시작해, 예측 비율과 목표 비율의 로그 차이의 절반(감쇠 0.5)씩 10번 더합니다. 클래스마다 같은 값을 더하므로 한 지문 안에서 클래스 간 순서를 바꾸는 효과는 "자주 나오는 클래스는 조금 덜, 드문 클래스는 조금 더"뿐입니다.

아래 셀은 \(\rho\)별 효과를 블렌드 OOF(2,699개)와 n-gram OOF(2,699개)에서 보여 줍니다. 두 집합 모두 클래스가 정확히 균등하므로 \(\rho\)가 클수록 유리하게 나옵니다. 그래서 여기서 \(\rho\)를 고르지 않고 `CFG["sinkhorn_rho"]` = 0.35를 그대로 씁니다. 테스트 지문이 200개 미만이면(인터랙티브 실행의 3행 등) 비율 추정이 불안정하므로 건너뜁니다.

````python
def sinkhorn_balance(lp, rho, n_iter, damping):
    """클래스별 로그 보정값을 더해 예측 비율을 (1-rho)*현재 + rho*균등 쪽으로 옮깁니다."""
    def softmax(z):
        z = z - z.max(1, keepdims=True)
        e = np.exp(z)
        return e / e.sum(1, keepdims=True)

    target = (1 - rho) * softmax(lp).mean(0) + rho / lp.shape[1]
    bias = np.zeros(lp.shape[1])
    for _ in range(n_iter):
        bias += damping * (np.log(target) - np.log(softmax(lp + bias).mean(0)))
    return lp + bias


def pred_spread(lp):
    counts = np.bincount(lp.argmax(1), minlength=K)
    return f"클래스별 예측 수 최소 {counts.min()} / 최대 {counts.max()}"


checks = {"블렌드 OOF": (eval_lp, y[eval_idx]), "n-gram OOF": (ng_oof_lp, y)}
for name, (lp, labels) in checks.items():
    print(f"[{name}] n={len(labels)}, 균등이면 클래스당 {len(labels) / K:.0f}개")
    for rho in (0.0, 0.35, 0.7, 1.0):
        out = lp if rho == 0 else sinkhorn_balance(lp, rho, CFG["sinkhorn_iters"], CFG["sinkhorn_damping"])
        mark = "  <- 사용" if rho == CFG["sinkhorn_rho"] else ""
        print(f"  rho={rho:<4}: BA {balanced_accuracy_score(labels, out.argmax(1)):.4f} | {pred_spread(out)}{mark}")

if CFG["use_sinkhorn"] and len(test) >= CFG["sinkhorn_min_rows"]:
    before = final_lp.argmax(1)
    final_lp = sinkhorn_balance(final_lp, CFG["sinkhorn_rho"], CFG["sinkhorn_iters"], CFG["sinkhorn_damping"])
    print(f"테스트에 적용: 예측이 바뀐 지문 {(final_lp.argmax(1) != before).sum()}개 / {len(test)}개 | {pred_spread(final_lp)}")
else:
    print(f"테스트 지문 {len(test)}개 < {CFG['sinkhorn_min_rows']}개라 테스트에는 적용하지 않습니다.")
````
